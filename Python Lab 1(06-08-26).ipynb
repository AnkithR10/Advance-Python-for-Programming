```python
a=list(map(int,input("Enter the list:").split()))
ele=int(input())
bol=False
for i in a:
    if i==ele:
        bol=True
        break
    else:
        bol=False
if bol:
    print("Found")
else:
    print("Not found")
```

```python
a=list(map(int,input("Enter the list:").split()))
n=len(a)
ele=int(input())
bol=False
for i in range(n):
    if a[i]==ele:
        bol=True
        break
    else:
        bol=False
if bol:
    print("Found")
else:
    print("Not found")
```

```python
#enumerate
color=["Blue","Red","White"]
for index,clr in enumerate(color):
    print(index,clr)
```

```python
#Zip function
name=["A","B","C"]
for n,clr in zip(name,color):
    print(n,clr)
```
