```python
#Closure 
def outer():
    message="Hello MCA"
    def inner():
        print(message)
    return inner

greet=outer()
greet()
print("Inspect closure:",greet)
print("Inspect using magic method:",greet.__closure__)
print("Cell content:",greet.__closure__[0].cell_contents)
```

```python
#Nested function
def process_marks(marks):
    def is_valid(mark):
        return 0<=mark<+100
    for mark in marks:
        if is_valid(mark):
            print(mark)

process_marks([85,90,105,76])
```

```python
#Student Marks Grade Calculator
def grade_cal(pass_marks):
    def calculate(marks):
        if marks>=pass_marks:
            return "Pass"
        else:
            return "Fail"
    return calculate

universal_res=grade_cal(40)
ug_grade=grade_cal(40)
pg_grade=grade_cal(50)

print("Ravi:",universal_res(40))
print("Raju:",ug_grade(35))
print("Ram:",pg_grade(80))
```

```python
#Why nonlocal is required?
def create_acc(initial_bal):
    balance=initial_bal
    def deposit(amount):
        nonlocal balance
        balance+=initial_bal
        return balance
    return deposit

account=create_acc(5000)
print(account(1000))
print(account(500))
```
