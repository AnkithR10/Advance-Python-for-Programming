```python
#Current directory
import os
print("Current Directory:",os.getcwd())
```

```python
#Create and list a directory
import os
folder="College"
if not os.path.exists(folder):
    os.mkdir(folder)
    print("Directory created")
else:
    print("Directory already exists")
print("Directory contents",os.listdir(folder))
```

```python
#Remove directory
import os
if os.path.exists(folder):
    os.rmdir(folder)
    print("Directory removed")
```

```python
#Create nested directory
import os
os.makedirs("College1/MCA/Semester1",exist_ok=True)
print(os.walk("Semester1"))
print("Nested directories created")
```

```python
#Rename a directory
import os

if os.path.isdir("College1"):
    os.rename("College1","NewRecord")
    print("DIrectory renamed")
else:
    print("Old records directory does not exist")
```

```python
import os
root_path='C:\Python Lab\College'
print(f"Walking through directory tree starting at:{root_path}\n")
for dirpath,dirnames,filenames in os.walk(root_path):
    print(f"Current Directory path:{dirpath}")
    print(f"Subdirectories inside it:{dirnames}")
    print(f"Files inside it:{filenames}")
    print("-" * 40)
```

```python
#Working with a file path
import os
path=os.path.join("Amrita","MCA","marks.csv")
if os.makedirs(path,exist_ok=True):
    print("Path created")
print("Path:",path)
print("File name:",os.path.basename(path))
print("Directory:",os.path.dirname(path))
print("Extension at 1:",os.path.splitext(path)[1])
print("Extension at 0:",os.path.splitext(path)[0])
print("Extension:",os.path.splitext(path))
print("Absolute path:",os.path.abspath(path))
```

```python
path="C:\Python Lab\Amrita\MCA\marks.csv"
if os.path.exists(path):
    if os.path.isfile("marks.csv"):
        print("File exist")
    if os.path.isdir("Amrita"):
        print("Directory exist")
```

```python
#pathlib module

from pathlib import Path
folder=Path("Amrita")
folder.mkdir(exist_ok=True)
print("Folder:",folder)
print("Exists:",folder.exists())
print("Is Directory:",folder.is_dir())
print("Is file:",folder.is_file())
print("Home directory:",folder.home())
print("Current directory:",folder.cwd())
print("Absolute path:",folder.resolve())
```

```python
from pathlib import Path
file=Path("C:\Python Lab\College\marks.csv.png")
print("File:",file)
print("File name:",file.name)
print("File extension:",file.suffix)
print("Multiple file extension:",file.suffixes)
print("Imediate directory:",file.parent)
print("Next Imediate directory:",file.parent.parent)
```
