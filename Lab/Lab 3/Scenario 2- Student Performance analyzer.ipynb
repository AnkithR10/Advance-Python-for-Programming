Scenario 2: Student Performance analyzer - *agrs

An MCA Student may study a diffrent Number of subjects depending on the semester.

The university wants a program that can calculate the student's percentage and grade regardless of the Number of subjects 


Problem Statements 

Develop a function using *args 

-accept any number of subjects

-calculate total and percentage

-Determines the Grade

-Returns the percentage

```python
from functools import reduce


def calculate(*marks):
    total_mark = reduce(lambda x,y : x+y ,marks,0)
    per = total_mark/len(marks)
    grade = ""
    match per:
        case x if x >95:
            grade = "O"
        case x if x  >85:
            grade = "A"
        case x if x  > 70:
            grade = "B"
        case x if x  >50:
            grade = "P"

    
    return per,grade

name = input("Name:").strip()
marks = list(map(int,input("Enter the Marks").split()))
percentage,grade = calculate(*marks)
print(f"Student name:{name}\nStudent Percentage:{percentage}\n Grade is {grade}")
#98,97,80,85,60
```
