```python
'''
An online store has a list of product Prices
The System Must:
-Add 18% GST to Every Product
-Considering product only greater than 500
-calculate te total using reduse()
'''
from functools import reduce
prices = list(map(float,input("Enter  the Product Price").split()))
#add 18%gst
price_with_gst = list(map(lambda x : x*1.18,prices))
filterd_price = list(filter(lambda x:x>=500,price_with_gst))
#Calculate Total
total = reduce(lambda x,y : x+y ,filterd_price,0)
print("Orginal Price",prices)
print("Price with GST",price_with_gst)
print("Products above 500:",filterd_price)
print("Total:",round(total,2))
```

```python
'''
Scenario 2: Student Performance analyzer - *agrs
An MCA Student may study a diffrent Number of subjects depending on the semester.
The university wants a program that can calculate the student's percentage and grade regardless of the Number of subjects 
Problem Statements 
Develop a function using *args 
-accept any number of subjects
-calculate total and percentage
-Determines the Grade
-Returns the percentage
'''
def calculate(*marks):
    total_mark = reduce(lambda x,y : x+y ,marks,0)
    per = total_mark/len(marks)
    grade = ""
    match per:
        case x if x >95:
            grade = "O"
        case x if x  >85:
            grade = "A"
        case x if x  > 70:
            grade = "B"
        case x if x  >50:
            grade = "P"

    
    return per,grade

name = input().strip()
marks = list(map(int,input("Enter the Marks").split()))
percentage,grade = calculate(*marks)
print(f"Student name:{name}\nStudent Percentage:{percentage}\n Grade is {grade}")
#98,97,80,85,60
```

```python
'''
home work
Electricity Bill calculator
An Electricity board wants to develop a pyton based billing system for diffrent categories of consumer
The Billing rules are:
cosumer_category Rate per-unit
Domestic         ₹5
Commercial       ₹8
Heavy Load       ₹12
Billing Rule:The first 120 units are free for evry consumer category
if consumption exceeds 120 units,the cosumer is charged only for the units cosumed beyond 120

'''
```

```python
category = input("Enter consumer category (Domestic/Commercial/Heavy Load): ").strip().lower()
units = float(input("Enter electricity units consumed: "))

rates = {
    "domestic": 5,
    "commercial": 8,
    "heavy load": 12
}

if category not in rates:
    print("Invalid consumer category.")
elif units < 0:
    print("Units cannot be negative.")
else:
    billable_units = max(units - 120, 0)
    bill = billable_units * rates[category]

    print(f"Consumer Category: {category.title()}")
    print(f"Units Consumed: {units:g}")
    print(f"Billable Units: {billable_units:g}")
    print(f"Electricity Bill: ₹{bill:.2f}")
```
