Electricity Bill calculator

An Electricity board wants to develop a pyton based billing system for diffrent categories of consumer

The Billing rules are:

cosumer_category Rate per-unit

Domestic         ₹5

Commercial       ₹8

Heavy Load       ₹12

Billing Rule:The first 120 units are free for evry consumer category

If consumption exceeds 120 units,the cosumer is charged only for the units cosumed beyond 120

```python
category = input("Enter consumer category (Domestic/Commercial/Heavy Load): ").strip().lower()
units = float(input("Enter electricity units consumed: "))

rates = {
    "domestic": 5,
    "commercial": 8,
    "heavy load": 12
}

if category not in rates:
    print("Invalid consumer category.")
elif units < 0:
    print("Units cannot be negative.")
else:
    billable_units = max(units - 120, 0)
    bill = billable_units * rates[category]

    print(f"Consumer Category: {category.title()}")
    print(f"Units Consumed: {units:g}")
    print(f"Billable Units: {billable_units:g}")
    print(f"Electricity Bill: ₹{bill:.2f}")
```
