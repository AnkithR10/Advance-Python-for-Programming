```python
#Iterators are objects that can be iterated upon, meaning that you can traverse through all the values. Technically, an iterator is any object which implements the iterator protocol, which consist of the methods __iter__() and __next__().
class EvenNumbers:
    def __init__(self,start,end):
        self.current = start
        self.end = end
        #Make sure the starting number is even
        if self.current % 2 != 0:
            self.current += 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.end:
            value = self.current
            self.current += 2
            return value
        raise StopIteration

obj=EvenNumbers(2, 10)
for number in obj:
    print(number)
```

```python
class EvenNumbers:
    def __init__(self,start,end):
        self.current = start
        self.end = end
        if self.current % 2 != 0:
            self.current += 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.end:
            value = self.current
            self.current += 2
            return value
        raise StopIteration

obj=EvenNumbers(2, 10)
print(next(obj))
print(next(obj))
print("\nUsing loop")
for number in obj:
    print(number)
```

```python
class Marks:
    def __init__(self, marks):
        self.marks = marks
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.marks):
            value = self.marks[self.index]
            self.index += 1
            return value
        raise StopIteration

    def total(self):
        return sum(self.marks)

    def average(self):
        return sum(self.marks) / len(self.marks)

student_marks = Marks([78,85, 91, 67, 88])
print("Display students marks with next() without loop:")
print(next(student_marks))
print(next(student_marks))
print(next(student_marks))
print(next(student_marks))
print(next(student_marks))

print("\nDisplay students marks with iterable object using loop:")
for mark in student_marks:
    print(mark)

print("\nTotal marks:", student_marks.total())
print("Average marks:", student_marks.average())
```

```python
#Generator for Squares:

def square_numbers(n):
    for i in range(n):
        yield i *i
print("Using loop:")
for number in square_numbers(5):
    print(number)
print("Using next():")
gen = square_numbers(5)
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
def prime_number(n):
    for num in range(2, n + 1):
        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            yield num
for num in prime_number(8):
    print(num)
```

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

for num in fibonacci(8):
    print(num)
```

```python
#Generator and lazy evaluation
def numbers():
    print("Starting generator")
    for i in range(1,4):
        print("Producing:", i)
        yield i

print("Creating generator")
g=numbers()

print("Generator created\n")
print("Requesting first value:")
print(next(g))

print("Requesting second value:")
print(next(g))
```

```python
from collection import defaultdict
students = [("Alice", "Computer Science"), ("Bob", "Mathematics"), ("Charlie", "Computer Science"), ("David", "Physics"), ("Eve", "Mathematics")]
departments=defaultdict(list)

for name,department in students:
    departments[department].append(name)

print("Students in each department:")
for department, student_list in departments.items():
    print(department,":",', '.join(student_list))
```
