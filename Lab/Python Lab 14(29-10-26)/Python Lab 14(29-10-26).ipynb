Report Generation

```python

import re
from collections import Counter
from datetime import datetime

# Match the timestamp, log level, and message
log_pattern = re.compile(
    r"^(\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2})"
    r"\s+(INFO|WARNING|ERROR|CRITICAL|DEBUG)\s+(.*)$"
)

counts = Counter()
timestamps = []
error_messages = []
total_entries = 0
invalid_entries = 0

# Read and analyze the log file
with open("application.log", "r", encoding="utf-8") as file:
    for line_number, line in enumerate(file, start=1):
        line = line.rstrip("\n")

        if not line.strip():
            continue

        match = log_pattern.fullmatch(line)

        if not match:
            invalid_entries += 1
            continue

        timestamp_text = match.group(1)
        level = match.group(2)
        message = match.group(3)

        try:
            timestamp = datetime.strptime(
                timestamp_text, "%Y-%m-%d %H:%M:%S"
            )
        except ValueError:
            invalid_entries += 1
            continue

        total_entries += 1
        counts[level] += 1
        timestamps.append(timestamp)

        if level in ("ERROR", "CRITICAL"):
            error_messages.append(
                f"{timestamp_text} {level} {message}"
            )

# Generate the report
with open("log_report.txt", "w", encoding="utf-8") as report:
    report.write("APPLICATION LOG ANALYSIS REPORT\n")
    report.write("=" * 40 + "\n\n")

    report.write(f"Total valid entries: {total_entries}\n")
    report.write(f"Invalid entries: {invalid_entries}\n\n")

    report.write("LOG LEVEL SUMMARY\n")
    report.write("=" * 40 + "\n")

    for level in ["INFO", "WARNING", "ERROR", "CRITICAL", "DEBUG"]:
        report.write(f"{level}: {counts[level]}\n")

    report.write("\nERROR AND CRITICAL MESSAGES\n")
    report.write("=" * 40 + "\n")

    if error_messages:
        for message in error_messages:
            report.write(message + "\n")
    else:
        report.write("No errors or critical events found.\n")

    report.write("\nTIMESTAMP SUMMARY\n")
    report.write("=" * 40 + "\n")

    if timestamps:
        report.write(
            f"Earliest timestamp: {min(timestamps)}\n"
        )
        report.write(
            f"Latest timestamp: {max(timestamps)}\n"
        )
    else:
        report.write("No valid timestamps found.\n")

print("Log analysis completed.")
print("Report saved as log_report.txt")
```

```python
#Fibonacci with stack memory
def Fibonacci(n):
    if n<=1:
        return n
    return Fibonacci(n-1)+Fibonacci(n-2)

print(Fibonacci(8))
```

```python
#Fibonacci usng built in Memoization
from functools import  lru_cache

@lru_cache(maxsize=None)
def Fibonacci(n):
    if n<=1:
        return n
    return Fibonacci(n-1)+Fibonacci(n-2)

n=int(input("Enter value to n:"))
for i in range(n):
    print(Fibonacci(i))

print("CAche information:",Fibonacci.cache_info())
```

```python
#Fibonacci using memoization
memo={}

def Fibonacci(n):
    if n in memo:
        return memo[n]
    if n<=1:
        return n
    memo[n]= Fibonacci(n-1)+Fibonacci(n-2)
    return memo[n]
n=int(input("ENter value of n:"))
for i in range(n+1):
    print(Fibonacci(i))
print(memo)
```

```python
#Word frequency
text="Python is easy and python is powerful and PYTHON has rich libraries"

words=text.lower().split()
freq={}
for word in words:
    freq[word]=freq.get(word,0)+1
print("Using loop:\n")

#Using Counter
from collections import Counter
freq=Counter(words)
print("\nUsing counter:\n",freq)
print("\nMost frequent word:\n",freq.most_common(2))

#Non Repeated words
flag=0
print("\nNon-repeating words:")
for word in words:
    if freq[word]==1:
        print(word)
        flag=1
if not flag:
    print("\nNo non-repeating words found")
```

```python
#Fixed size sliding window
#Maximum subarray sum 
arr=[2,4,6,8,10,]
k=3
w_sum=sum(arr[:k])
max_sum=w_sum
start_index=0
for i in range(k,len(arr)):
    w_sum=w_sum-arr[i-k]+arr[i]
    if w_sum>max_sum:
        max_sum=w_sum
        start_index=i-k+1
max_subarry=arr[start_index:start_index+k]
print("Maximum sum:",max_sum)
print("Elements contributing to maximu sum:",max_subarry)
```
