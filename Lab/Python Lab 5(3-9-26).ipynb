```python
try:
    a=int(input("Enter a value to a:"))
    b=int(input("ENter value to b:"))
    print("Result:",a/b)
except ValueError:
    print("Please enter valid integer values")

except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Done")
```

```python
#Handle multiple exceptions together
try:
    number=int(input("Enter a number:"))
    result=100/number
    print(result)

except(ValueError,ZeroDivisionError) as error:
    print("Error:",error)
```

```python
#File handling error
try:
    f=open("student.txt")
    try:
        f.write("Lorum Ipsum")
    except:
        print("Something went wrong when writing to the file (first)")
    finally:
        f.close()
except:
    print("Something went wrong when writing to the file(second)")
```

```python
#Raise-Explicitly raising an exception
age=int(input("Enter age:"))
if age<0:
    raise ValueError("Age cannot be negative")
print("Age=",age)
```

```python
#Raise exceptin if marks not between 0 and 100
marks=int(input("Enter marks:"))
if marks<0 or marks>100:
    raise ValueError("Marks must be between 0 and 100")
```

```python
#Assert statement
#If the condition is false, an AssertionError is raised
age=20
assert age>18
print("Eligible")
```

```python
def calculate_average(*mark):
    assert len(mark)>0,"Marks list cannot be empty"
    return sum(mark)/len(mark)
print("Average marks:",calculate_average(87,92))
print("Average marks:",calculate_average(87,92,99))
print("Average marks:",calculate_average())
```

```python
#User defined exception
class InvalidMarks(Exception):
    pass
Marks=int(input("Enter marks: "))
try:
    if Marks<0 or Marks>100:
        raise InvalidMarks("Marks must be between 0 and 100")
    print("Marks accepted:",Marks)#60,123
except InvalidMarks as error:
    print("Error:",error)
```
