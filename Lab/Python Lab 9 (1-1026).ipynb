# Decorators

```python
def make_multiplier(factor):
    def multipler(number):
        return number * factor
    return multipler

double=make_multiplier(2)
triple=make_multiplier(3)
print("Double:",double)
print("Tripple:",triple)

print("Double:",double(5))
print("Triple:",triple(5))

print("Inspect the double memory:",double.__closure__[0].cell_contents)
print("Inspect the triple memory:",triple.__closure__[0].cell_contents)
```

```python
#Building a Decorator manually
def decorator(function):
    def wrapper():
        print("Before function execution")
        function()
        print("After function execution")
    return wrapper

def display():
    print("Hello")
display=decorator(display)
display()
```

```python
#Building a Decorator manually
def decorator(function):
    def wrapper():
        print("Before function execution")
        function()
        print("After function execution")
    return wrapper
@decorator
def display():
    print("Hello")
display()
```

```python
#Decorator with function argument

def decorator(function):
    def wrapper(a,b):
        print("Addition started")
        result=function(a,b)
        print("Addition completed")
        return result
    return wrapper

@decorator
def add(a,b):
    return a+b

print(add(10,20))
```

```python
# Generic wrapper using *args and **kwargs
def decorator(function):
    def wrapper(*args, **kwargs):
        print("Function name:", function.__name__)
        result = function(*args, **kwargs)
        print("Execution completed")
        return result
    return wrapper

@decorator
def add(a, b):
    return a + b

print(add(10, 20))
```

```python
import time
def timer(function):
    def wrapper(*args,**kwargs):
        start=time.time()
        result=function(*args,**kwargs)
        end=time.time()
        print("Execution Time:",end-start,"seconds")
        return result
    return wrapper

@timer
def calculate_sum():
        total=0
        for i in range(1000000):
            total+=i
        return total
print("Result:",calculate_sum())
```

A banking app. requires:
- Authentication
- Logging


Create two decorators

```python
#Multiple deorator
def authenticate(function):
    def wrapper(user,*args,**kwargs):
        if user=="admin":
            return function(user,*args,**kwargs)
        print("Authentication failed")
    return wrapper

def log(function):
    def wrapper(*args,**kwargs):
        print("Transaction started")
        result=function(*args,**kwargs)
        print("Transaction complete")
        return result
    return wrapper

@authenticate
@log
def transfer_money(user,amount):
    print("Transfered $",amount)


transfer_money("admin",5000)
transfer_money("student",5000)
```

```python
from functools import wraps
def retry(times):
    def decorator(function):
        @wraps(function)
        def wrapper(*args,**kwargs):
            for attempt in range(1,times+1):
                try:
                    return function(*args,**kwargs)
                except Exception as e:
                    print("Attempt",attempt,"failed:",e)
            print("All attemptsfailed")
        return wrapper
    return decorator

@retry(3)
def connect_server():
    print("Connecting to server....")
    raise ConnectionError("Server unavailable")

connect_server()
```
