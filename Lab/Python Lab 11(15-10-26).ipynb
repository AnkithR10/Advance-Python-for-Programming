```python
#Introspection
class Student:
    name="Sudhi"
    def display():
        pass

student=Student()
field=input("Enter attribute name:")
value=input("Enter attribute value:")
setattr(student,field,value)
print("Updated value:",getattr(student,field))
print(f"Student has attribute{field}:",hasattr(student,field))

for i in dir(Student):
    if not i.startswith("__"):
        print(i)

if hasattr(student,field):
        delattr(student,field)
        print(f"Attribute '{field}'deleted successfully")

print(f"Student has attribute {field}:",hasattr(student,field))

print("Methods defined:")
for attr in dir(student):
    if not attr.startswith("__"):
        print(attr)

print("Type of student:",type(student))
print("Student callable:",callable(Student))
print("Display callable:",callable(display))
print("Methods available",dir(student))
```

```python
#Line by line reading
search_id=input("Enter record ID:").strip()
found=False

with open("large.txt","r",encoding="utf-8") as file:
    for line in file:
        line=line.strip()

        record_id=line.split(":")[0].split()[1]
        if record_id==search_id:
            print("Record found:",record_id)
            found=True
            break

if not found:
    print("Record not found")
```

# Byte streams in python
- Bytes are immutable sequence of bytes
- bytearry is a mutable sequence of bytes
- Byte I/O

```python
data=bytes([65,66,67])
print("Bytes:",data)

mutable_dataarray=bytearray([65,66,67])
mutable_dataarray[0]=90
print("Modified bytearray:",mutable_dataarray)
print("As text:",mutable_dataarray.decode("ascii"))
#data[0]=90
#print("Modified bytes:".data)
```

```python
#Byte I/P 
from io import BytesIO
stream=BytesIO()
stream.write(b"Hello MCA")
print("Position after wrinting:",stream.tell())
stream.seek(0)
data=stream.read()
print("Data:",data)
print("getvalue():",stream.getvalue())
print("As text:",data.decode("utf-8"))
```

# Serialization and Deserialization
- Serialization converts an object into representation that can be stored or transmitted
- Deserialization reconstructs the object from representation

```python
import pickle
import os
filename="student_records.pkl"
def load_records():
    if os.path.exists(filename):
        with open(filename,"rb") as file:
            return pickle.load(file)
    return []

def save_records(records):
    with open(filename,"wb")as file:
        pickle.dump(records,file)
    
def add_student(records):
    roll_no=input("Enter roll number:").strip()
    if any(s["roll_no"]==roll_no for s in records):
        print("Roll number already exists")
        return
    name=input("Enter name:").strip()
    marks=float(input("Enter marks:"))
    records.append({"roll_no":roll_no,"name":name,"marks":marks})
    save_records(records)
    print("Student record saved")

def display_students(records):
    if not records:
        print("No records available")
        return
    for student in records:
        print(student)

def search_student(records):
    roll_no=input("Enter roll number to search:").strip()
    for student in records:
        if student["roll_no"]==roll_no:
            print("Record found:",student)
            return
    print("Student not found")

records=load_records()
while True:
    print("\n1. Add student")
    print("2. AddDisplay student")
    print("3. Search student")
    print("Exit")

    choice=input("Enter your choice:").strip()

    if choice=="1":
        add_student(records)
    elif choice=="2":
        display_students(records)
    elif choice=="3":
        search_student(records)
    elif choice=="4":
        print("Exiting program")
        break
    else:
        print("Invalid choice")
```
