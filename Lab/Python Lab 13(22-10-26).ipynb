```python
#Calculate the total size of files
import os
total_size=0
for root,dirs,files in os.walk("C:\Python Lab\College"):
    for file in files:
        path=os.path.join(root,file)
        try:
            total_size +=os.path.getsize(path)
        except OSError:
            print("Could not access:",path)
print("Total size:",total_size,"bytes")
```

```python
#Using json.dumps() and json.loads()
import json
student={"name":"Lavanya","marks":90,"passed":True}
json_text=json.dumps(student)
print("JSON data:",json_text)

restored_student=json.loads(json_text)
print("Student name:",restored_student["name"])
print("Marks:",restored_student["marks"])
```

```python
#Writing JSON to a file and reading it back
import json
student={"roll_no":102,"name":"Ravi","marks":[80,85,90]}

#Write JSON to a file
with open("student.json","w",encoding="utf-8") as file:
    json.dump(student,file,indent=4)
print("Student data saved")

#Read JSON from the file
with open("student.json","r",encoding="utf-8") as file:
    data=json.load(file)
print("Student:",data["name"])
print("Total marks:",sum(data["marks"]))
```

```python
#handling invalid JSON
import json
text='{"name":"Anu","marks":90}'
try:
    data=json.loads(text)
    print(data)
except json.JSONDecodeError as error:
    print("Invalid JSON:",error)
```

```python
#Update the existing JSON file
import json
filename="student_report.json"
with open(filename,"r",encoding="utf-8")as file:
    students=json.load(file)
for student in students:
    python_marks=float(student["Python_marks"])
    dbms_marks=float(student["DBMS_marks"])
    student["Total"]=python_marks+dbms_marks
    student["Average"]=student["Total"]/2
    with open(filename,"w",encoding="utf-8")as file:
        json.dump(student,file,indent=4)

print("Existing JSON file updated successfully")
```

```python
import csv
filename="results.csv"
with open(filename,"r",newline="",encoding="utf-8") as file:
    reader=csv.DictReader(file)
    students=list(reader)
    fields=reader.fieldnames.copy()
for column in ["Total","Average"]:
    if column not in fields:
        fields.append(column)

for student in students:
    python_marks=float(student["python_marks"])
    dbms_marks=float(student["dbms_marks"])
    total=python_marks+dbms_marks
    average=total/2
    student["Total"]=total
    student["Average"]=average

with open(filename,"w",newline="",encoding="utf-8") as file:
    writer=csv.DictWriter(file,fieldnames=fields)
    writer.writeheader()
    writer.writerows(students)
print("Existing CSV file updated successfully")
```
