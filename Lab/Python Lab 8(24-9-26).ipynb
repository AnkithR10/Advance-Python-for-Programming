```python
class Counter:
    def __init__(self,start,end):
        self.current=start
        self.end=end
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.current<=self.end:
            value=self.current
            self.current+=1
            return value
        raise StopIteration

    
counter=Counter(1,5)

print("With iterable abject using loop")
for number in counter:
    print(number)

print("\nUsing next() manually instad of loop")
counter1=Counter(10,15)
print(next(counter1))
print(next(counter1))
print(next(counter1))
```

```python
#Generator
def counter(start,end):
    for i in range(start,end+1):
        yield i

for num in counter(1,5):
    print(num)
```

```python
#Generator and lazy evaluation(DONT WRITE IN OBSERVATION)
def number():
    print("Starting generator\n")
    for i in range(1,4):
        print("Producing:",i)
        yield i

print("Creating generaor")
g=number()
print("Generator created\n")
print("Requesting first value\n")
print(next(g))

print("Requesting second value\n")
print(next(g))
```

```python
#A server has million log rcords
def read_log(logs):
    for log in logs:
        yield log

def extract_status(logs):
    for log in logs:
        timestap,user_id,status=log.split(",")
        yield timestap,user_id,int(status)

def filter_errors(records):
    for timestap,user_id,status in records:
        if status>=400:
            yield timestap,user_id,status

logs=["10:01,U101,200","10:02,U102,404","10:03,U103,200","10:04,U104,500","10:05,U105,505"]
data=read_log(logs)
data=extract_status(data)
errors=filter_errors(data)
count=0

for error in errors:
    print(error)
    count+=1
print("Total Errors:",count)
```

```python
#Modules
#Collections
#Counter
from collections import Counter
sentence="python data science python machine learning data python"
words=sentence.split()
frequency=Counter(words)
print("Freq. of words:",frequency)
print("Display the most common words:",frequency.most_common(2))
```

```python
#Defaultdict
from collections import defaultdict
students=[("Anita","CSE"),("Rahul","ECE"),("Priya","CSE"),("Arjun","ISE"),("Meena","ECE")]
departments=defaultdict(list)

for name,department in students:
    departments[department].append(name)

for department,name in departments.items():
    print(department,":",name)
```

```python
#Functools
#Reduce
from functools import reduce
numbers=[2,3,4,5]
result=reduce(lambda x,y:x*y,numbers)
print("Product:",result)
```

```python
#Partial
from functools import partial
def calculate_price(price,tax_rate):
    return price+(price*tax_rate/100)
gst_18=partial(calculate_price,tax_rate=18)
print("Laptop price:",gst_18(50000))
print("Mobile price:",gst_18(20000))
```

```python
#Cycle
from itertools import cycle
lights=cycle(["RED","GREEN","YELLOW"])

for i in range(8):
    print(next(lights))
```

```python
#Chain
from itertools import chain
cse=["Bharath","Prajwal"]
ece=["Pranav","Akash"]
ise=["Ankith","Bhanuprasad"]
students=chain(cse,ece,ise)
for st in students:
    print(st)
```

```python
#Combination
from itertools import combinations
students=["Prajwal","Bharath","Akash","Pranav"]
groups=combinations(students,3)
for group in groups:
    print(group)
```

```python
#Permutations
from itertools import permutations
students=["A","B","C"]
for arrange in permutations(students):
    print(arrange)
```

```python
#Product
from itertools import product

sizes=["S","M","L"]
colors=["Red","Blue"]
for i in product(sizes,colors):
    print(i)
```

```python
#Math
#Calculate the distance between (2,3) and (7,9)
import math as m
x1,y1=2,3
x2,y2=7,9
euc=m.sqrt((x2-x1)**2 +(y2-y1)**2)
city=abs((x2-x1)+(y2-y1))
chess=max(abs(x2-x1),abs(y2-y1))
print("Euclidean Distance:",euc)
print("Manhattan/City block distance:",city)
print("Minkowaski/Chess bound distance:",chess)
```
