```python
#Short hand if
age=int(input("Enter your age: "))
if age>=18: print("You are eligible to vote")  
```

```python
#Short hand if-else
age=int(input("Enter your age: "))
print("You are eligible to vote") if age>=18 else print("You are not eligible to vote")
```

```python
#Assiign value with short hand if-else
sal1=int(input("Enter salary:"))
sal2=int(input("Enter salary:"))
high= sal1 if sal1>sal2 else sal2
print(high)
```

```python
#Multiple statements in one line
age=int(input("Enter your age: "))
if age>=18: print("You are eligible to vote"); print("You can also contest elections")
```

```python
#Combining multiple Operators
age=int(input("Enter your age: "))
if age>=18 and age<=60: print("You are eligible to vote"); print("You can also contest elections")
```

```python
#Leap year using short hand if-else
year=int(input("Enter a year: "))
print("Leap year") if (year%4==0 and year%100!=0) or (year%400==0) else print("Not a leap year")
```

```python
#Using parantheses for Clarity
temp=25
is_rain=False
is_weekend=True
if(temp>20 and not is_rain)or is_weekend:
    print("Great day for outdoor activities!")
```

```python
#Multiple conditions
marks=int(input("Enter marks:"))
print("O") if marks>= 90 else\
print("A+") if marks>=80 else\
print("A") if marks>=70 else\
print("B+")if marks>=60 else\
print("F")
```

```python
# Fibonacci series using while
n = int(input("Enter number of terms: "))

n1, n2 = 0, 1
i = 0

while i < n:
    print(n1)
    n1, n2 = n2, n1 + n2
    i += 1
```

```python
#Sorting the list in ascending order
numbers=[5,2,8,1]
for num in sorted(numbers):
    print(num)
```

```python
#Reverse using while loop
count=5
while count>0:
    print(count)
    count-=1
    
```
