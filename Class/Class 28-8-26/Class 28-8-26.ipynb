Methods:

        write()
        
        open()
        
        close()

```python
#Create/Open file
file=open("student.txt","w")
print("File created successfully")
file.close()
```

```python
#Writing to a file
file=open("student.txt","w")
file.write("Ankith\n")
file.write("Hello\n")
file.write("Welcome\n")
file.close()
```

```python
#Writing multiple lines using writelines()
students=["Ankith\n","Bharath\n","Pranav\n"]
file=open("student.txt","w")
file.writelines(students)
file.close()
```

```python
#Appending to a file
students=["Ankith\n","Bharath\n","Pranav\n"]
file=open("student.txt","a")
file.writelines(students)
file.close()
```

```python
#Reading a complete file
file=open("student.txt","r")
content=file.read()
print(content)
file.close()
```

```python
#Reading a specific number of characters
file=open("student.txt","r")
content=file.read(10)
print(content)
file.close()
```

```python
#Reading a specific number of characters
file=open("student.txt","r")
content=file.read(20)
print(content)
file.close()
```

```python
#Reading one line at a time
file=open("student.txt","r")
line1=file.readline()
print(line1)
file.close()
```

```python
#Reading all lines into a list
file=open("student.txt","r")
lines=file.readlines() #Notice that readlines() returns a list of strings
print(lines)
file.close()
```

Reading lines using loops

```python
# Reading lines using a loop
with open("student.txt", "r") as file:
    for line in file:
        print(line, end="")
```
