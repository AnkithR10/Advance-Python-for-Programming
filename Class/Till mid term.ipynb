```python
'''
CONSTRUCTOR: A special method automatically invoked when a new object is created. 
Its primary purpose is to initialize the object's variables and set up its initial state.

TYPES: 
- Default constructor
- Non-parameterized Constructor
- Parameterized Constructor
'''

# default constructor

class EmptyClass:
    name = "Arun"
    course = "MCA"
    age = 20
    
    @staticmethod   # A static method performs an operation that does not require object or class data.
    def funMethod(): # method without any parameter
        print("Roopa is pursuing BCA, and his age is 18")

    def funMethods(self):  # Non-parameterized method
        print(f"{self.name} is pursuing {self.course}, and his age is {self.age}")

# Python uses a default constructor to create the object
obj = EmptyClass() 
print(obj.name)
print(obj.course)
print(obj.age)
obj.funMethod()
obj.funMethods()
```

```python
'''
Static Methods
Methods that are completely independent of both the class and individual objects. 
They behave like regular utility functions but are visually nested inside 
the class namespace for logical grouping. 
They cannot access instance or class variables directly.
'''

class Security:

    @staticmethod
    def validate_password(password):
        return len(password) >= 8

password = input("Enter password: ")
print("Length of the password valid: ", Security.validate_password(password))
```

```python
'''
Class Methods
A class method is useful when the information belongs to the class as a whole.

Methods that are bound to the class itself rather than an individual object. 
They can access and modify class-level variables (variables shared by all instances) 
and accept the class type as their first parameter.
'''

class Ride:
    base_fare = 50

    @classmethod
    def change_base_fare(cls, new_fare):
        cls.base_fare = new_fare

print("Base-fare: ", Ride.base_fare)
Ride.change_base_fare(60)
print("Changed-fare: ", Ride.base_fare)
```

```python
'''
Static method vs Class method
- Class method → works with class-level data
- Static method → performs a utility operation related to the class independently, but does not need class or object data

| Feature                         | Class Method           | Static Method             |
| ------------------------------  | ---------------------- | ------------------------- |
| Decorator                       | `@classmethod`         | `@staticmethod`           |
| First parameter                 | `cls`                  | None                      |
| Access object data directly     | No                     | No                        |
| Access class data directly      | Yes                    | No                        |
| Modify class variables          | Yes                    | No                        |
| Depends on object               | No                     | No                        |
| Depends on class                | Usually yes            | No                        |
| Main purpose                    | Class-level operations | Utility/helper operations |
'''

class Dog:
    @staticmethod
    def info():
        print("Dogs are loyal animals.")

    @classmethod
    def count(cls):
        print("There are many dogs of class", cls)

Dog.info()  # Static method call
Dog.count()  # Class method call
```

```python
'''
Class Variables vs Instance Variables
Class Variables are shared by all instances of a class. 
They are defined inside the class but outside any methods.
Instance Variables are unique to each instance and are defined in the init() method.
'''

class Pets:
    species = "Canine"  # Class variable

    def __init__(self, name, age):
        self.name = name  # Instance variable
        self.age = age  # Instance variable

dog1 = Pets("Buddy", 3)
dog2 = Pets("Lucy", 2)

print("Species of pet: " + dog1.species)  # Accessing class variable
print("Pets name: " + dog1.name + f" and is {dog1.age} years old")  # Accessing instance variable
print("Pets name: " + dog2.name + f" and is {dog2.age} years old")  # Accessing instance variable
```

```python
'''
Methods 
Functions defined inside a class that describe the behaviors or actions an object can perform. 
They typically manipulate the object's instance variables
'''

class Student:
    def __init__(self, name, marks): # parameterized method
        self.name = name
        self.marks = marks

    def get_info(self):             # non-parameterized method
        return f"{self.name} is {self.marks} years old"

    def display(self):              # non-parameterized method
        print("Name:", self.name)
        print("Marks:", self.marks)

    # __str__() method is a special method that controls what is returned when the object is printed
    def __str__(self):
        return f"Person name: {self.name}, her age is {self.marks} years."  

stud = Student("Anita", 85)         # stud object invokes __init__() by default constructor
stud.display()                      # stud object invokes non-parameterized method
print(stud.get_info())
print(stud)                         # stud object is getting printed

# del stud.display                   # the del keyword is used to delete methods from a class
```

```python
'''
Encapsulation - The practice of bundling data (variables) and methods together 
into a single unit (a class) while restricting direct outside access. 
This is usually achieved by making variables private and 
exposing them only through controlled public get and set methods. 

Encapsulation provides several benefits:
- Data Protection: Prevents accidental modification of data
- Validation: You can validate data before setting it
- Flexibility: Internal implementation can change without affecting external code
- Control: You have full control over how data is accessed and modified
'''

class DigitalWallet:
    def __init__(self, balance):
        self.__balance = balance

    def add_money(self, amount):
        if amount > 0:
            self.__balance += amount

    def spend_money(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient wallet balance")

    def get_balance(self):
        return self.__balance

wallet = DigitalWallet(2000)
wallet.add_money(500)
wallet.spend_money(300)

print("Balance:", wallet.get_balance())
```

```python
'''
Polymorphism
Meaning "many forms," it lets a single interface, method, or object behave
differently depending on the context or the type of object acting on it.
'''

class EmailNotification:
    def send(self):
        print("Sending email notification")

class SMSNotification:
    def send(self):
        print("Sending SMS notification")

class PushNotification:
    def send(self):
        print("Sending push notification")

# one object invoking multiple constructors
notifications = [EmailNotification(), SMSNotification(), PushNotification()]  

for notification in notifications:
    notification.send()
```

```python
'''
Operator overloading allows a user-defined class to give special meaning 
to operators such as +, -, *, ==, <, >, etc.

Arithmetic Operators
| Operator   | Magic Method     |
| ---------- | ---------------- |
| `+`        | `__add__()`      |
| `-`        | `__sub__()`      |
| `*`        | `__mul__()`      |
| `/`        | `__truediv__()`  |
| `//`       | `__floordiv__()` |
| `%`        | `__mod__()`      |
| `**`       | `__pow__()`      |
| `@`        | `__matmul__()`   |
| `-obj`     | `__neg__()`      |
| `+obj`     | `__pos__()`      |
| `abs(obj)` | `__abs__()`      |

'''

# Suppose a financial system has two investment accounts. 
# Adding two account objects should produce their combined balance.

class InvestmentAccount:
    def __init__(self, account_no, balance):
        self.account_no = account_no
        self.balance = balance

    def __add__(self, other):
        return self.balance + other.balance

a1 = InvestmentAccount("INV101", 75000)
a2 = InvestmentAccount("INV102", 45000)

total = a1 + a2

print("Combined Investment:", total)
```

```python
import math

class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading the + operator
    def __add__(self, other):
        # self is the left operand, other is the right operand
        return Vector2D(self.x + other.x, self.y + other.y)

    # Overloading the < operator (Less Than)
    def __lt__(self, other):
        # Compare vectors based on their magnitude (length)
        self_mag = math.sqrt(self.x**2 + self.y**2)
        other_mag = math.sqrt(other.x**2 + other.y**2)
        print(f"{self_mag} < {other_mag}")
        return self_mag < other_mag

    # Overloading the string representation for clean printing
    def __str__(self):
        return f"Vector({self.x}, {self.y})"

# Create two vector objects
v1 = Vector2D(3, 4)   
v2 = Vector2D(1, 2)   

v3 = v1 + v2   # Use the overloaded + operator
print(f"{v1} + {v2} = {v3}\n")  # printing object with the help of __str__()

# Use the overloaded < operator
print(f"Is v1 < v2? {v1 < v2}\n")  
print(f"Is v2 < v1? {v2 < v1}") 
```

```python
'''
Inheritance
A mechanism that allows a new class (subclass/child) to acquire 
the properties and methods of an existing class (superclass/parent). 
This promotes code reusability and builds hierarchical relationships.
'''

# Single Inheritance: One child class inherits from one parent class.

class Vehicle:
    def start_engine(self):
        print("Engine started")

class Car(Vehicle):
    def drive(self):
        print("Car is ready go on drive")

car = Car()

car.start_engine()
car.drive()
```

```python
# Multiple Inheritance: One child class inherits from more than one parent class.

class Camera:
    def take_photo(self):
        print("Photo captured")

class GPS:
    def locate(self):
        print("Current location identified")

class SmartPhone(Camera, GPS):
    def make_call(self):
        print("Calling...")

phone = SmartPhone()

phone.take_photo()
phone.locate()
phone.make_call()
```

```python
# Hybrid Inheritance: A combination of two or more types of inheritance.

class Device:
    def power_on(self):
        print("Device powered on")

class Camera(Device):
    def capture(self):
        print("Image captured")

class GPS(Device):
    def locate(self):
        print("Location identified")

class SmartDevice(Camera, GPS):
    def connect(self):
        print("Smart device connected")

device = SmartDevice()

device.power_on()
device.capture()
device.locate()
device.connect()

print(SmartDevice.mro())
```

```python
# Method Resolution Order (MRO) – When multiple inheritance is used, Python needs to determine which method should be executed first.

class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        print("B")

class C(A):
    def show(self):
        print("C")

class D(B, C):
    print("D")

obj = D()

obj.show()
print(D.mro())
```

```python
'''
2. Object Representation
__str__() and __repr__(): These methods control how an object is represented as text.

Research Paper Management: Create a ResearchPaper class with: title, author, year
Use __str__() for user-friendly output, __repr__() for developer/debugging representation
'''

class ResearchPaper:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def __str__(self):
        return f"{self.title} - {self.author}"

    def __repr__(self):
        return (f"ResearchPaper("f"'{self.title}', "f"'{self.author}', "f"{self.year})")

paper = ResearchPaper("AI in Autonomous solutions","Dr. Manohar",2026)

print(paper)
print(repr(paper))
```

```python
'''
3. Container / Collection Behavior
__len__(), __contains__(), __getitem__() and __setitem__()
These make a user-defined object behave like a built-in collection.

Digital Library: Create a Library class that stores book titles.
Use len(library) to find the number of books.
'''

class Library:
    def __init__(self, books):
        self.books = books

    def __len__(self):
        return len(self.books)

    def __contains__(self, book):
        return book in self.books

library = Library(["Python", "Data Mining", "Machine Learning", "Computer Vision"])

print("Number of books:", len(library))

print("Python in library:", "Python" in library)
print("Java in library:", "Java" in library)
```

```python
'''
4. Indexing and Item Assignment

Student Marks Manager: Create a Marks class that allows students' marks to be accessed and modified using indexing.
The following should work: marks[0], marks[1] = 95
Validate that marks are between 0 and 100.
'''

class Marks:
    def __init__(self, marks):
        self.marks = marks

    def __getitem__(self, index):
        return self.marks[index]

    def __setitem__(self, index, value):
        if value < 0 or value > 100:
            raise ValueError("Marks must be between 0 and 100")
        self.marks[index] = value

marks = Marks([85, 90, 78, 88])

print("First mark:", marks[0])
print("orginal third mark:", marks[2])
marks[2] = 95
print("Updated third mark:", marks[2])
```

```python
'''
5. Arithmetic Operator Overloading

Create a Product class; calculate the total price for purchased items with the discount applied.
'''

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __mul__(self, quantity):
        return self.price * quantity

    def __sub__(self, discount):
        return self.price - discount.amount

class Discount:
    def __init__(self, amount):
        self.amount = amount

itemName = input("Enter the name of the item: ")
price = int(input("Enter product price: "))
quantity = int(input("Enter quantity: "))

product = Product(itemName, price)

# Determine discount
if price < 1000:
    discount = Discount(0)

elif price < 5000:
    discount = Discount(100)

elif price < 8000:
    discount = Discount(500)

else:
    discount = Discount(3000)


total = product * quantity   # object * variable

price_aftr_dscnt = product - discount   # object of class Product * object of class Discount

final_price = price_aftr_dscnt * quantity  

print(f"Total for {quantity} {itemName}: ₹{total}")
print("Discount per item: ₹", discount.amount)
print("Price after discount: ₹", final_price)
```

```python
'''
6. Comparison magic methods: __eq__()   → ==; __ne__()   → !=, __lt__()   → <, __gt__()   → >, __le__()   → <=, __ge__()   → >=

Student CGPA Comparison: A university wants to compare the CGPA of two students. 
The Student class should support comparison magic methods.
'''

class Product:
    def __init__(self, name, price, rating, reviews, discount, warranty):
        self.name = name
        self.price = price
        self.rating = rating
        self.reviews = reviews
        self.discount = discount
        self.warranty = warranty

    def score(self):
        review_score = min(self.reviews / 100, 10)
        overall_score = (self.rating * 20 + self.discount * 2 + self.warranty * 5 + review_score)
        return review_score, overall_score
    
    def __str__(self):
        review_score, overall_score = self.score()
        return f"{review_score}"
    
    def __eq__(self, other):   # ==
        return self.score()[0] == other.score()[0]
   
    def __ne__(self, other):   # !=
        return self.score()[1] != other.score()[1]

    def __lt__(self, other):    # <
        return self.score() < other.score()

    def __gt__(self, other):    # >
        return self.score() > other.score()

    def __le__(self, other):    # <=
        return self.score()[1] < other.score()[0] or self.reviews == other.reviews

    def __ge__(self, other):    # >=
        return self.score()[0] >= other.score()[1]

p1 = Product("Laptop A", 55000, 4.5, 1200, 10, 2)
p2 = Product("Laptop B", 60000, 4.7, 850, 15, 3)

print("Laptop A review score, overall score:", p1.score())
print("Laptop B review score, overall score:", p2.score())

print("\nOverall Scores")
review1, score1 = p1.score()
review2, score2 = p2.score()

print("Laptop A overall score:", score1)
print("Laptop B overall score:", score2)

print("\nReview Scores")
print(f"Review Score of {p1.name}:", p1)
print(f"Review Score of {p2.name}:", p2)

print("\nComparison")
print(f"{p1.score()[0]} == {p2.score()[0]}:", p1 == p2)
print(f"{p1.score()[1]} != {p2.score()[1]}:", p1 != p2)
print(f"{p1} < {p2}:", p1 < p2)
print(f"{p1} > {p2}:", p1 > p2)
print(f"{p1.score()[1]} <= {p2.score()[0]}:", p1 <= p2)
print(f"{p1.score()[0]} >= {p2.score()[1]}:", p1 >= p2)
```

```python
'''
Unary Operators: __neg__(), __pos__(), __abs__(), __invert__()
'''
# Sensor Error Analyzer: A sensor may report a positive or negative measurement error.

class SensorError:
    def __init__(self, error):
        self.error = error

    def __neg__(self):       # Unary -
        return SensorError(-self.error)

    def __pos__(self):    # Unary +
        return SensorError(+self.error)

    def __abs__(self):    # Absolute value
        return abs(self.error)

    def __invert__(self):     # Bitwise NOT
        return SensorError(~self.error)

    def __str__(self):
        return f"Sensor Error: {self.error}"

error = SensorError(7)

print("Original:", error)
print("Unary - :", -error)
print("Unary + :", +error)
print("Absolute:", abs(error))
print("Bitwise ~:", ~error)
```

```python
'''
Matrix Operator: @ access __matmul__()
'''

class Matrix:
    def __init__(self, data):
        self.data = data

    def __matmul__(self, other):
        result = []
        for i in range(len(self.data)):
            row = []
            for j in range(len(other.data[0])):
                total = 0
                for k in range(len(other.data)):
                    total += (self.data[i][k] * other.data[k][j])
                row.append(total)
            result.append(row)
        return Matrix(result)

    def display(self):
        for row in self.data:
            print(row)

A = Matrix([[1, 2],[3, 4]])
B = Matrix([[5, 6],[7, 8]])
C = A @ B
print("matrix multiplication")
C.display()
```

```python
'''
Callable Objects: __call__() It allows an object to be called like a function.
'''

class AIModel:
    def __init__(self, threshold):
        self.threshold = threshold
    def __call__(self, score):
        if score >= self.threshold:
            return f"{score} >= {self.threshold}: Pass"
        return f"{score} < {self.threshold}: Fail"

model = AIModel(50)

print(model(75))
print(model(35))
```

```python
'''
Create a ShoppingCart class that supports: 
number of products, check product existence, 
access product, combine carts, compare carts and display cart
'''

class ShoppingCart:
    def __init__(self, products):
        self.products = products

    def __len__(self):
        return len(self.products)

    def __contains__(self, product):
        return product in self.products

    def __getitem__(self, index):
        return self.products[index]

    def __add__(self, other):
        return ShoppingCart(self.products + other.products)

    def __eq__(self, other):
        return self.products == other.products

    def __str__(self):
        return f"Shopping Cart: {self.products}"

cart1 = ShoppingCart(["Laptop", "Mouse"])
cart2 = ShoppingCart(["Keyboard", "Headset"])
cart3 = cart1 + cart2
print(cart3)
print("\nNumber of products:", len(cart3))
print("\nMouse is in cart:" , "Mouse" in cart3)
print("\nFirst product:", cart3[0])
cart4 = ShoppingCart(["Laptop", "Mouse"])
print("\ncart1 == cart4:", cart1 == cart4)
```
