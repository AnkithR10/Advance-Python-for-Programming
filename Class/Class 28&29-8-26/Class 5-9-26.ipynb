Summary: Debugging Techniques in Python
Debugging means finding and fixing errors in a program. Common Python debugging techniques include:
- Print statements/logging: Display variable values and program flow to understand what the program is doing.
- Exception handling: Use try, except, and finally to handle runtime errors safely and identify the cause.
- Assertions: Use assert to check whether an expected condition is true. It helps verify program correctness during development.
- Breakpoints: Use breakpoint() to pause the program at a specific line and inspect variable values and execution flow.
- Call stack: The call stack shows the sequence of function calls that led to the current point or an error. It helps trace where a problem originated.
- Python debugger: Python includes a built-in debugger called pdb. Calling breakpoint() usually starts this debugger.


Debuging is the process of identifying, analyzing and correcting errors(bugs) in a program.
It is an essential programming skill because real world software may fail due to:
- Incorrect program logic
- Invalid user input
- Runtime exceptions

```python
#Assert
def calculate_average(numbers):
    assert len(numbers) > 0, "The list of numbers must not be empty."
    return sum(numbers) / len(numbers)

numbers = [10, 20, -10,30,-1]
calculate_average(numbers)
# numbers_empty = []
# empty_average = calculate_average(numbers_empty)
```

### Logging
- Level-Purpose
- Debug-Detailed info for debugging
- Info-
- Debug
- Warning
- Critical - Indicates serious problem

```python
import logging
logging.basicConfig(level=logging.DEBUG, format="%(levelname)s:%(message)s")

logging.debug("Program started")

age=int(input("Enter your age: "))
logging.debug(f"User entered age: {age}")

if age<18:
    logging.warning("Age is below eligibility criteria")

logging.debug("Program completed")
```

4. Python debugger(pdb)

Python provides a built-in debugger called pdb.

It allows the programmer to:
- Pause program execution
- Execute

Commands:
- n- execut next line
- s- step into a function
- c - continue execution
- p variable- print variable
- l
- q- To quit
- w- Display calls of the debugger

```python
import pdb
marks=[80,90,75]
total=sum(marks)
pdb.set_trace()  
average=total/len(marks)
print("Average marks:", average)
```
