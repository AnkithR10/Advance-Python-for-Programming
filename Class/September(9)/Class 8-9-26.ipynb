Scenario: Smart Traffic Light Simulator
Imagine you are developing a Smart Traffic Management System for a busy city.
The system should simulate traffic signals at different road intersections and automatically change their colours.
For each intersection, use a dictionary to represent two traffic directions:
- NS represents the North–South traffic direction.
- EW represents the East–West traffic direction.
Each direction can have one of the following signal colours:
- Red
- Yellow
- Green
For example, the traffic light status at Market Street can be represented as:
market_street = {
    "NS": "Green",
    "EW": "Red"
}

```python
market_2nd={'ns':'green', 'ew':'red'}
mission_16th={'ns':'red', 'ew':'green'}
def switch(stoplight):
    for key in stoplight.keys():
        if stoplight[key]=='green':
            stoplight[key]='yellow'
        elif stoplight[key]=='yellow':
            stoplight[key]='red'
        elif stoplight[key]=='red':
            stoplight[key]='green'

    assert 'red' in stoplight.values(), \
        'Neither light is red! '+str(stoplight)

switch(mission_16th)
print("Traffic light status:",mission_16th)
switch(market_2nd)
print("Traffic light status:",market_2nd)
```
