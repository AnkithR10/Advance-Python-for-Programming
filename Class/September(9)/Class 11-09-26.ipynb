```python
#An online food delivery application
#Creates an order usg customer name,food item, and quantity

class FoodOrder:
    def __init__(self, customer,item, quantity):
        self.customer = customer
        self.item = item
        self.quantity = quantity
order1=FoodOrder("Anita", "Pizza", 1)
order2=FoodOrder("Bharath","Dosa", 2)

print("Customer Name:", order1.customer)
print("Food Item:", order1.item)
print("Quantity:", order1.quantity)

print("Customer Name:", order2.customer)
print("Food Item:", order2.item)
print("Quantity:", order2.quantity)
```
