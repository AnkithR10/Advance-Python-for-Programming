# Decorators
So what is the decorator? It's also a function. It's also a function OK which modifies the behavior of another function without altering the. Logic or the behavior of the original function Like as we say. Biotech worker, but which means I'm not touching or modifying the original user defined function using the original user defined function, I'm just adding on additional functionality, additional features Or adding or modifying. OK you're understanding that is said to be a function which adds or modifies certain behavior of the existing function. 

```python
def casechange(function):
    def innerfun():
        return function().upper()
    return innerfun

@casechange
def message():
    return "namah shivaya"

print(message())
```

Function as an object

```python
def greet():
    print("Hello")

message=greet
message()
```

Function as an arguement

```python
def greet():
    print("Hello")

def execute(function):
    function()

execute(greet)
```

Nested Function

```python
def outer():
    def inner():
        print("Inside inner function")
    inner()
outer()
```

```python
def outer(name):
    def inner():
        print("Hello",name)
    inner()
outer("Lahari")
```
