```python
#Suppose a student stands 50 m away from a building and observes the top at an angle of elevation 40 degree.Using:tan(0)=distance/height
import math as m
dis=50
a=40
radian=m.radians(a)
height=dis*m.tan(radian)
print("Heifht of building:",height,"meters")
```

```python
#Suppose a drone travels 100m at an angle 30 above the horizontal.Find its vertical displacement
distance=100
angle=30
radian=m.radians(angle)
vertical_dis=distance*m.sin(radian)
print("Vertical distance:",vertical_dis,"meters")
```

# Date and time module

```python
from datetime import datetime as d
now=d.now()

print("Current date and time:",now)
print(now.strftime("%d-%m-%Y"))
print(now.strftime("%d/%m/%Y"))
print(now.strftime("%B-%d-%Y"))
```

```python
from datetime import date
dob=date(2000,8,15)
today=date.today()
age=today.year- dob.year
if (today.month,today.day)<(dob.month,dob.day):
    age-=1
print("Age:",age)
```

```python
from datetime import datetime,timedelta,date
todays=date.today()
now =datetime.now()
future_date=todays+timedelta(days=7)
print("Today:",todays)
print("After 7 days:",future_date)

past_date=now-timedelta(hours=3,minutes=30)
print("3.5 hours ago:",past_date)
```
