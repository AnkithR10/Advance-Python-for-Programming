Accept a list of values for marks

If marks is not a number handle it with exception handling

```python
#marks=[25,35,90,"AB",87,"Pranav"]
marks=map(list,input("Enter the marks separated by space: ").split())
updated=[]
for i in marks:
    try:
        updated.append(int(i))
    except ValueError:
        print(f"Invalid mark {i} found.")

print("Updated marks list:", updated)
```

```python
#Bank traction using exception handling
amount=10000
try:
    withdraw=float(input("Enter the amount to withdraw:"))
    if withdraw>amount:
        raise ValueError("Insufficient funds.")
    elif withdraw<=0:
        raise ValueError("Invalid withdrawal amount.")
    else:
        amount-=withdraw
        print(f"Transaction successful. Remaining balance: {amount}")
except ValueError as e:
    print(e)
```

Write a Python program for a Student Record Management System with the following requirements:
1. Create a user-defined exception named InvalidMarksError.
2. Accept student name and marks from the user.
3. The student name must not be empty or contain only whitespace.
4. Marks must be numeric and must lie between 0 and 100.
5. Raise InvalidMarksError when marks are non-numeric, negative, or greater than 100.
6. Use exception handling to display appropriate error messages.
7. Store only valid student records in a text file named student_records.txt.
8. Allow the user to enter multiple student records.
9. Stop the program when the user enters exit as the student name.

```python
class InvalidMarksError(Exception):
    """Raised when marks are non-numeric or outside 0 to 100."""
    pass


def validate_marks(marks_text):
    try:
        marks = float(marks_text)
    except:
        raise InvalidMarksError("Marks must be numeric.")

    if marks < 0 or marks > 100:
        raise InvalidMarksError("Marks must be between 0 and 100.")

    return marks


file_name = "student_records.txt"

print("Student Record Management System")
print("Enter 'exit' as the student name to stop.\n")

while True:
    try:
        name = input("Enter student name: ").strip()

        if name.lower() == "exit":
            print("Program ended.")
            break

        if name == "":
            raise Exception("Student name cannot be empty.")

        marks_input = input("Enter marks (0 to 100): ").strip()
        marks = validate_marks(marks_input)

        with open(file_name, "a") as file:
            file.write(f"Name: {name}, Marks: {marks}\n")

        print("Student record saved successfully.\n")

    except InvalidMarksError as error:
        print("Invalid Marks Error:", error)
        print("Record was not saved. Try again.\n")

    except Exception as error:
        print("Error:", error)
        print("Record was not saved. Try again.\n")
```
