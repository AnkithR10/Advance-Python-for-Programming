Accept a list of values for marks

If marks is not a number handle it with exception handling

```python
#marks=[25,35,90,"AB",87,"Pranav"]
marks=map(list,input("Enter the marks separated by space: ").split())
updated=[]
for i in marks:
    try:
        updated.append(int(i))
    except ValueError:
        print(f"Invalid mark {i} found.")

print("Updated marks list:", updated)
```

```python
#Bank traction using exception handling
amount=10000
try:
    withdraw=float(input("Enter the amount to withdraw:"))
    if withdraw>amount:
        raise ValueError("Insufficient funds.")
    elif withdraw<=0:
        raise ValueError("Invalid withdrawal amount.")
    else:
        amount-=withdraw
        print(f"Transaction successful. Remaining balance: {amount}")
except ValueError as e:
    print(e)
```
