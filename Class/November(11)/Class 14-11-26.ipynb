# Unit testing with unittest module

- unittest module
    - python -m unittest  [filename] - for single file test
    - python -m unittest discover -s  [starting_name] - for all test files at once
- Test suites:
    combination multiple module of test cases into one execution

- unittest Fixtures:
    - setUp - at the begining before all other testcases 
    - tearDown - executes at last as clearing things

- Mock Objects: A mock replaces an external dependency during testing, such as:
    - Database
    - API call
    - File system
    - Payment gateway
    - Email service
    - Current time

## pytest
- Uses assert command 
- simple compared to unittesting
- runs all testable files automatically

It automatically detects the followinhg:
    - test

- To all test  in a file - syntax: pytest [foldername]/[filename] -v
- To single class test in a file - syntax: pytest [foldername]/[filename]/[functionname]
