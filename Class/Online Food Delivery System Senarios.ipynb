You are developing an Online Food Delivery System.

When a customer places an order, the system should automatically assign the nearest available delivery partner to minimise delivery time. If all delivery partners are busy, the order should be placed in a waiting queue until a partner becomes available.

Suppose there is one restaurant and four delivery partners.

Restaurant Location: (x, y)

Business Rules

Only available delivery partners are eligible.

Select the partner who has completed the fewest deliveries today (fair workload distribution).

If multiple partners have the same number of deliveries, select the partner with the highest customer rating.

If there is still a tie, select the nearest delivery partner.

If there is still a tie, select the partner who has been idle the longest.

Once assigned: (a) Update the partner's status to Busy, (b) Increment the number of deliveries completed today.

```python
partners = [
    ["D101", "ABC", 12, 11, 1, 8, 5],
    ["D102", "PQR", 9, 8, 1, 4, 4],
    ["D103", "XYZ", 9, 8, 0, 4, 3],
    ["D104", "DEF", 11, 9, 1, 2, 5]
]
restaurant = (6, 7)
available = [p for p in partners if p[4] == 1]
available.sort(key=lambda p: (p[5], -p[6], (p[2] - restaurant[0]) ** 2 + (p[3] - restaurant[1]) ** 2))
selected = available[0]
selected[4] = 0
selected[5] += 1
print("Delivery Assigned to", selected[0])
```
