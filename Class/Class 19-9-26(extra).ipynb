Keyword Arguments

Parameter name while calling the function

key=value syntax

The position does not matter here

Variable length keyword arguments- **kwargs

```python
#Student details using Keyword Arguments

def student(name,age,course):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)

student(course="MCA", name="Ankith", age=22)#Keyword Arguments
print()
student("Ankith", 22, "MCA") #Positional Arguments
```

Variable length Keyword Arguments- **kwargs

**kwargs allows a function to accept any number of keyword arguments

```python
def student_details(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

student_details(name="Ankith", age=22, course="MCA", semester=3)  # Keyword Arguments with variable-length
```

```python
# Positional and Keyword Arguments with variable-length
def student(*subjects, **details):
    print("Subjects:", subjects)
    for key, value in details.items():
        print(f"{key}: {value}")

student("Python","DBMS","Machine Learning",name="Ankith",semester=3)  
```

Return Values

Functions can send data back  to the code that called them using the return statement.
When a function reaches a return statement, it stops executing and sends the result

Returning Multiple values
Python function can return multiple values


```python
#Difference between print() and return()
def add_print(a, b):
    print(a + b)

def add_return(a, b):
    return a + b

res1=add_print(5, 10) 
print("Result from add_print:", res1)

res2=add_return(5, 10)
print("Result from add_return:", res2)

#The first function displays the result but returna None.
#The second function returns the result, which can be stored in a variable and used later.
```

```python
#Returning multiple values from a function
def result(*marks):
    total=sum(marks)
    percentage=(total/len(marks))
    return total, percentage

total,percentage=result(80, 90, 85, 95)
print("Total:", total)
print("Percentage:", percentage)
```

```python
#Returning multiple values from a function
def result(marks):
    total=sum(marks)
    percentage=(total/len(marks))
    return total, percentage

marks=list(map(int, input("Enter marks separated by space: ").split()))
total,percentage=result(marks)
print("Total:", total)
print("Percentage:", percentage)
```

Scope of Variables

Scope determines where a variable can be accessed:

Local and Global

```python
#Global keyword
balance = 1000  # Global variable

def deposit(amount):
    global balance
    balance += amount

deposit(500)
print("Balance:",balance)
```

Recursion

Recursion is when a function calls itself

This has the benefit of meaning that you can loop through data to reach a result.

```python
#Factorial using recursion
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

n=int(input("Enter a number to calculate its factorial: "))
result=factorial(n)
print(f"Factorial of {n} is: {result}")
```

```python
#Tail recursion= The recursive call is the last thing the function does

def factorial(n,result=1):
    if n == 0:
        return result
    else:
        return factorial(n-1, n*result)

print(factorial(1000))
```

```python
#Sum of n numbers using tail recursion
def sum_n(n, result=0):
    if n == 0:
        return result
    else:
        return sum_n(n-1, n+result)

print(sum_n(2000))
```

```python
#Fibonacci series using tail recursion
def fibonacci(n, a=0, b=1):
    if n == 0:
        return a
    elif n == 1:
        return b
    else:
        return fibonacci(n-1, b, a+b)

print(fibonacci(2000))
```
