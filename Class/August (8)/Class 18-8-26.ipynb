Positional only Arguments- To specify position value only seperated by ,/

```python
#function with positional arguments
def parameter_function(pet, name):
    print("I have a", pet)
    print("My", pet+"'s name is", name)

def my_function(name,/):
    print("name:", name)

parameter_function("dog", "Buddy")
parameter_function("Buddy", "dog")
my_function("Ankith")
```

```python
# Sum of n elements using variable-length positional arguments
def calculate_sum(*numbers):
    total = 0
    for number in numbers:
        total += number
    return total

print(calculate_sum(10, 20))
print(calculate_sum(10, 20, 30, 40))
```
