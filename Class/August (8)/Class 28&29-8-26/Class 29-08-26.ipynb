```python
#Different methods to read a file
file = open("student.txt", "r+")
readcontent = file.read()
file.write("\nThank You")

print("readcontent:", readcontent)
print("File name:", file.name)
print("File mode:", file.mode)
print("File closed:", file.closed)
```

```python
#the with statement automatically closes the file after the nested block of code
with open("student.txt", "a") as file:
    name=input("Enter your name: ")
    file.write(name +"\n")
print("Student added successfully")
print("File closed:", file.closed)
```

```python
#Counting the number of lines in a file
with open("student.txt", "r") as file:
    content=file.read()

lines=content.splitlines()
words=content.split()
characters=len(content)
print("Number of lines:", len(lines))
print("Number of words:", len(words))
print("Number of characters:", characters)
```

```python
#Counting the number of lines,words and charaters in a file using a loop
with open("student.txt", "r") as file:
    line_count=0
    word_count=0
    char_count=0
    for line in file:
        line_count+=1
        word_count+=len(line.split())
        char_count+=len(line)

print("Number of lines:",line_count)
print("Number of words:",word_count)
print("Number of characters:",char_count)
```

```python
#The file pointer tell() indicates thr current position in the file
with open("student.txt", "r") as file:
    print("Initial position:",file.tell())
    print(file.read(5))
    print("Position after reading:",file.tell())
```

```python
#Tell() with write mode
with open("student.txt", "w") as file:
    print("Initial position:",file.tell())
    file.write("Hello World")
    print("Position after writing:",file.tell())
```

```python
#seek() moves the file pointer to a specific position
with open("student.txt","r") as file:
    print(file.read(10))
    file.seek(5)
    print("Position after seeking:",file.tell())
    print("Current text:",file.read(6))
```
