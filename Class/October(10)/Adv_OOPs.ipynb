```python
'''
Iterable Objects:
An iterable is an object whose elements can be accessed one by one.

The iter() function obtains an iterator from an iterable.

Iterable
   ↓ iter()
Iterator
   ↓ next()
Individual elements

Iterable vs Iterator
---------------------
- The Iterable (The Container): Holds the data/configuration and returns 
a fresh iterator every time __iter__() is called.

- The Iterator (The State Tracker): Holds the changing state of a single traversal.

To put it simply: the iterable is the book, 
and the iterator is the bookmark.

Why Iterable vs Iterator?
------------------------------
--- Iterable ---
- An iterable usually holds all its elements in memory 
at once (like a list of 1 million integers).

- An iterable (like a list or tuple) is a data container. 
It knows what data it holds, but it doesn't know—and shouldn't care—where you currently are in a loop.

- Use an Iterable when you want a permanent container that you can 
loop through multiple times.

--- Iterator ---
- An iterator does not need to store the data. 
It only needs to know the rule to calculate the next item when you call next(). 
This allows you to stream massive files, process infinite sequences, or 
fetch data from a database one row at a time without crashing your computer's memory.

- An iterator is a temporary pointer. 
It tracks the exact current position (state) and knows how to fetch the __next__() item.

- Use an Iterator when you need a temporary stream that remembers 
its current position and fetches data on demand.

Iterator Protocol
---------------------
Python follows an iterator protocol for sequential traversal.
An iterator should provide two methods:__iter__() and __next__()

__iter__(): Returns the iterator object.
__next__(): Returns the next value.

for loop
   ↓
iter(object)
   ↓
iterator
   ↓
next(iterator)
   ↓
value
   ↓
next(iterator)
   ↓
value
   ↓
...
   ↓
StopIteration
'''
```

```python
class CountdownContainer:
    """The Iterable: Holds the data blueprint and can be looped multiple times."""
    def __init__(self, start):
        self.start = start

    def __iter__(self):
        # Always returns a BRAND NEW iterator instance
        return CountdownIterator(self.start)


class CountdownIterator:
    """The Iterator: Tracks the state of one specific loop execution."""
    def __init__(self, count):
        self.count = count

    def __iter__(self):
        return self

    def __next__(self):
        if self.count <= 0:
            raise StopIteration
        current = self.count
        self.count -= 1
        return current

# --- How to use it ---
timer = CountdownContainer(4)
time = CountdownIterator(4)

print("Display students marks initially with next(): ")
print(next(time))
print(next(time))

print("\nFirst loop:")
for num in timer: print(num)  

print("\nSecond loop (it works again!):")
for num in timer: print(num)  

print("\nDisplay students marks later with next(): ")
print(next(time))
print(next(time))
print(next(time))
```

```python
# Create an EvenNumbers iterator that generates even numbers between a start and end value.

class EvenNumbers:
    def __init__(self, start, end):
        self.current = start
        self.end = end
        # Make sure starting value is even
        if self.current % 2 != 0:
            self.current += 1
    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= self.end:
            value = self.current
            self.current += 2
            return value
        raise StopIteration

obj = EvenNumbers(2, 10)

print("using next()")
print(next(obj))
print(next(obj))

print("\nusing loop")
for number in obj:
    print(number)
```

```python
# Create a StudentMarks class that returns marks one at a time and also displays the total and average.

class Marks:
    def __init__(self, marks):
        self.marks = marks
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index < len(self.marks):
            value = self.marks[self.index]
            self.index += 1
            return value
        raise StopIteration
    def total(self):
        return sum(self.marks)
    def average(self):
        return sum(self.marks) / len(self.marks)
student_marks = Marks([78, 85, 91, 67, 88])
print("Display students marks with next() without using loop: ")
# print(next(student_marks))
# print(next(student_marks))
# print(next(student_marks))
# print(next(student_marks))
#print(next(student_marks))

print("\nDisplay students marks with iterable object using loop: ")
for mark in student_marks:
    print(mark)

print("\nTotal Marks:", student_marks.total())
print("Average:", student_marks.average())
```

```python
# Custom Iterator

class Counter:
    def __init__(self, start, end):
        self.current = start
        self.end = end

    def __iter__(self):  # Always returns a BRAND NEW iterator instance
        return self      # Why __iter__() Returns self - This object itself is the iterator.

    def __next__(self):
        if self.current <= self.end:
            value = self.current
            self.current += 1
            return value
        raise StopIteration

counter = Counter(1, 5)

# with iterable object using loop
print("With iterable object using loop")
for number in counter:
    print(number)

# Using next() manually instead of loop
print("\nUsing next() manually instead of loop")
counter1 = Counter(10, 15)
print(next(counter1))
print(next(counter1))
print(next(counter1))
```

```python
'''
Generators
---------------------------------------------------------------
Generators are special tools used to create iterators in a 
highly memory-efficient manner. Instead of computing an 
entire dataset and storing it in a computer's RAM all at once, 
generators produce values one at a time, only when requested
—a concept known as lazy evaluation.

A generator function looks exactly like a normal function, 
but it uses the yield keyword instead of return.

How it works: A regular function runs to completion and loses its state. 
When a generator function is called, it returns a generator object 
without running the code immediately. Every time you call next() on 
that object (or loop over it), the function runs until it hits yield. 

It then pauses, preserves its entire state, and outputs the value. 
When you call it again, it resumes exactly where it left off.
```

```python
# Generator

def counter(start, end):
    for i in range(start, end + 1):
        yield i

for number in counter(1, 5):
    print(number)
```

```python
# Generator for Squares: Create a generator that produces squares from 1 to n.

def square_numbers(n):
    for i in range(1, n + 1):
        yield i * i
print("using loop:")
for number in square_numbers(5):
    print(number)
print("using next():")
gen = square_numbers(5)
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
# Prime Number Generator: Generate prime numbers up to n without first creating a list of primes.

def prime_numbers(n):
    for number in range(2, n + 1):
        is_prime = True
        for i in range(2, int(number ** 0.5) + 1):
            if number % i == 0:
                is_prime = False
                break
        if is_prime:
            yield number

for prime in prime_numbers(20):
    print(prime)
```

```python
# Fibonacci Generator: Generate the first n Fibonacci numbers.

def fibonacci(n):
    a = 0
    b = 1
    for i in range(n):
        yield a
        a, b = b, a + b

for number in fibonacci(8):
    print(number)
```

```python
# A server has millions of log records.
def read_logs(logs):
    for log in logs:
        yield log

def extract_status(logs):
    for log in logs:
        timestamp, user_id, status = log.split(",")
        yield timestamp, user_id, int(status)

def filter_errors(records):
    for timestamp, user_id, status in records:
        if status >= 400:
            yield timestamp, user_id, status

logs = ["10:01,U101,200", "10:02,U102,404", "10:03,U103,200", "10:04,U104,500", "10:05,U105,403"]
data = read_logs(logs)
data = extract_status(data)
errors = filter_errors(data)
count = 0

for error in errors:
    print(error)
    count += 1
print("Total Errors:", count)
```

```python
# A data science application wants to measure how long a function takes to execute.

import time

def timer(function):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = function(*args, **kwargs)
        end = time.time()
        print("Execution Time:", end - start, "seconds")
        return result
    return wrapper

@timer
def calculate_sum():
    total = 0
    for i in range(1000000):
        total += i
    return total

print("Result:", calculate_sum())
print("Result:", calculate_sum())
```

```python
'''
functools.wraps
The metadata can refer to the wrapper rather than the original function.

without @wrap()
1. Python looks at calculate and sees the wrapper function instead. 
Printing __name__ would output "wrapper", and __doc__ would output None
(because wrapper has no docstring). 
This breaks debugging tools and auto-generated documentation.

With @wraps(): 
The @wraps decorator acts as a copy-paste tool for metadata. 
It copies the original function's name, docstring, module, and other 
properties directly onto the wrapper function before returning it.
'''  

from functools import wraps

def decorator(function):
    #@wraps(function)
    def wrapper(*args, **kwargs):
        return function(*args, **kwargs)
    return wrapper

@decorator
def calculate():
    """Calculate student result."""
    pass

print(calculate.__name__)
print(calculate.__doc__)
```

```python
# Create a decorator that prevents a result from being displayed if the student's marks are below 40.

from functools import wraps

def check_pass(function):
    @wraps(function)
    def wrapper(name, marks):
        if marks < 40:
            print(name, "has failed")
            return
        return function(name, marks)
    return wrapper

@check_pass
def display_result(name, marks):
    print(name, "has passed with", marks)

display_result("Shivu", 85)
display_result("Ramya", 35)
```

```python
# Decorator Factory: decorator that accepts an argument

def repeat(times):
    def decorator(function):
        @wraps(function)
        def wrapper(*args, **kwargs):
            for i in range(times):
                function(*args, **kwargs)
        return wrapper
    return decorator

@repeat(3)
def greet(name):
    print("Bonjour", name)

greet("Raghu")
```

```python
'''
You are developing a university REST API.
Different users have different permissions:
Admin   → Add/Delete/View
Faculty → Add/View
Student → View
that checks whether the logged-in user's role has permission to execute the function.
'''

from functools import wraps
def permission(required_role):
    def decorator(function):
        @wraps(function)
        def wrapper(user_role, *args, **kwargs):
            if user_role == required_role:
                return function(user_role,*args,**kwargs)
            print("Permission denied")
        return wrapper
    return decorator

@permission("admin")
def delete_student(role, student_id):
    print("Student", student_id, "deleted")

delete_student("admin", 101)
delete_student("student", 101)
```

```python
# A machine-learning application contains several expensive functions

import time
from functools import wraps

def timer(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        start = time.perf_counter() # measure execution time
        result = function(*args, **kwargs)
        end = time.perf_counter()
        print(function.__name__, "took", end - start, "seconds")
        return result
    return wrapper

@timer
def preprocess_data():
    total = 0
    for i in range(10000):
        total += i
    return total


@timer
def extract_features():
    total = 0
    for i in range(5000):
        total += i
    return total

preprocess_data()
extract_features()
```

```python
''' A network application occasionally fails while connecting to a server.
create @retry(3) which attempts the function a maximum of three times. '''

from functools import wraps

def retry(times):
    def decorator(function):
        @wraps(function)
        def wrapper(*args, **kwargs):
            for attempt in range(1, times + 1):
                try:
                    return function(*args, **kwargs)
                except Exception as e:
                    print("Attempt",attempt,"failed:",e)
            print("All attempts failed")
        return wrapper
    return decorator

@retry(3)
def connect_server():
    print("Connecting to server...")
    raise ConnectionError("Server unavailable")

connect_server()
```

```python
'''
Context Manager
A context manager is a Python mechanism used to manage resources automatically. 
It ensures that a resource is properly acquired before use and properly released 
after use, even if an error occurs.

Common resources include:
- Files
- Database connections
- Network connections
- Locks
- Temporary resources
- Threads/process-related resources
The most common way to use a context manager is the with statement.

Resource Management
---------------------
Whenever a program uses a resource, two operations are generally required:
Acquire Resource
       ↓
   Use Resource
       ↓
Release Resource
---------------------
For example, when working with a file:
Open file
   ↓
Read/write file
   ↓
Close file

The general syntax is:
with resource as variable:    # use resource

Example
--------------------
with open("students.txt", "r") as file:
    content = file.read()
    print(content)

Why Is with Better?
---------------------
Consider:
file = open("students.txt")
try:    
    data = file.read()
finally:    
    file.close()

This manually handles cleanup.
---------------------------------
The with version:
with open("students.txt") as file:    
    data = file.read()

is shorter and safer.
---------------------------------
Conceptually:
with
 ↓
Acquire resource
 ↓
Execute block
 ↓
Cleanup automatically
---------------------------------
Even if an exception occurs:
with open("students.txt") as file:    
    data = file.read()    
    x = 10 / 0

Python still performs the required cleanup.
'''
```

```python
'''
User-Defined Context Managers:
Python allows us to create our own context managers.

A class-based context manager generally implements:
__enter__()
__exit__()
These are special methods.

__enter__() Method
__enter__() is executed when Python enters the with block.

Syntax:
def __enter__(self):
...


__exit__() Method
The __exit__() method is called when the with block is exited.

Syntax:
def __exit__(self, exc_type, exc_value, traceback):
...

The three parameters provide information about an exception, if one occurred.
Parameter	| Meaning
---------------------------------------
exc_type	| Type of exception
exc_value	| Exception object/message
traceback	| Traceback information
'''
```

```python
class Demo:
    def __enter__(self):
        print("Entering context")
        return self
    def __exit__(self, exc_type, exc_value, traceback):
        print("Exiting context")

with Demo():
    print("Inside with block")
```

```python
'''
What Does __enter__() Return?
We can use the value returned by __enter__() with as.
'''
class Student:
    def __enter__(self):
        print("Entering")
        return "MCA Student"
    def __exit__(self, exc_type, exc_value, traceback):
        print("Exiting")

with Student() as value:
    print(value)
```

```python
# __exit__() With Exception
class Demo:
    def __enter__(self):
        print("Start")
    def __exit__(self, exc_type, exc_value, traceback):
        print("Exception:", exc_value) # __exit__() is still called.
        print("Cleanup completed") # If __exit__() returns False or None, the exception normally propagates.

with Demo():
    print("Processing...")
    x = 10 / 0
```

```python
'''
Suppressing an Exception
__exit__() can also determine whether an exception should be suppressed.
'''

class SafeOperation:
    def __enter__(self):
        print("Operation started")
    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type:
            print("Exception handled:", exc_value)
            return True

with SafeOperation():
    print("Performing operation")
    x = 10 / 0
print("Program continues")
```

```python
# Database Transaction Using a Class-based Context Manager 

class DatabaseTransaction:
    def __enter__(self):
        print("Transaction Started")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type is None:
            print("Transaction Completed")
        else:
            print("Transaction Failed")
            print("Exception:", exc_value)
            return True  # Suppress the exception


print("Example 1: Normal Execution") 
with DatabaseTransaction():
    print("Inserting student record")
    print("Updating student marks")

print("\nExample 2: Exception Handling")
with DatabaseTransaction():
    print("Inserting student record")
    x = 10 / 0
    print("Updating student marks")

print("Program continues normally")
```

```python
class FileLogger:
    def __enter__(self):
        self.file = open("application.log", "a")
        self.file.write("Log session started\n")
        print("Log session started")
        return self.file

    def __exit__(self, exc_type, exc_value, traceback):
        self.file.write("Log session ended\n")
        self.file.close()
        print("Log session ended")

with FileLogger() as log:
    log.write("Student logged in\n")
    print("Student logged in")
    log.write("Student viewed marks\n")
    print("Student viewed marks")

print("File closed successfully")
```

```python
# Student Lab Resource Manager Using a Class-Based Context Manager
class LabResourceManager:
    def __init__(self, student_name):
        self.student_name = student_name
        self.computer = "Computer-MCA2601"

    def __enter__(self):
        print("Computer allocated to", self.student_name)
        print("Allocated computer:", self.computer)
        return self.computer

    def __exit__(self, exc_type, exc_value, traceback):
        print("Computer released")
        if exc_type is not None:
            print("Error occurred:", exc_value)
            print("Exception handled successfully")
            return True
        return False

# Example 1: Normal execution
print("=== Normal Lab Session ===")

with LabResourceManager("Sree") as computer:
    print("Sree is using", computer)
    print("Executing Python program")

# Example 2: Exception occurs
print("\n=== Lab Session with Exception ===")

with LabResourceManager("Roshin") as computer:
    print("Roshin is using", computer)
    result = 10 / 0

print("Program continues normally")
```

```python
# contextlib for Resource Management

from contextlib import contextmanager

@contextmanager
def database_connection():
    print("Database connection opened")
    connection = "Database Connection Object"
    try:
        yield connection
    finally:
        print("Database connection closed")

with database_connection() as db:
    print("Using:", db)
```

```python
'''
Duck Typing
-------------------
Duck typing means that Python focuses on what an object can do 
rather than requiring the object to belong to a particular class.

The idea is often summarized as:
If it behaves like a duck, treat it like a duck.

For example, if an object supports a required method such as start(), 
a function can use that method without checking the object's class first.
```

```python
class Car:
    def start(self):
        print("Car engine started")

class Bike:
    def start(self):
        print("Bike engine started")

class Computer:
    def start(self):
        print("Computer started")

def start_device(device): # does not check whether device is a Car, Bike, or Computer
    device.start() # It simply expects the object to provide a callable start() method.

car = Car()
bike = Bike()
computer = Computer()

start_device(car)
start_device(bike)
start_device(computer)
```

```python
'''
Notice that isinstance(True, int) returns True in Python 
because bool is a subclass of int. If Boolean values must 
be rejected, an explicit check is needed:

if isinstance(marks, (int, float)) and not isinstance(marks, bool):
    print("Valid numeric type")
'''

def calculate_result(marks):
    if isinstance(marks, (int, float)) and not isinstance(marks, bool):
        print("Valid numeric type")
    else:
        print("Invalid marks type")

calculate_result(85)
calculate_result(92.5)
calculate_result(False)
calculate_result("Absent")
```

```python
# Type checking with inheritance

class Person:
    pass

class Student(Person):
    pass

s = Student()
print("---isinstance()---")
print(isinstance(s, Student))
print(isinstance(s, Person))
print(issubclass(Student, Person))

'''
isinstance() recognizes the inheritance relationship, 
whereas type(s) == Person would be False because the 
actual type of s is Student.
'''

print("---type()---")
print(type(s) == Student)
print(type(s) == Person)
```

```python
'''
Mutable vs. Immutable Objects
Python objects can be classified according to whether their 
contents can be changed after creation.

- Mutable objects: Their contents can be modified after creation.
- Immutable objects: Their contents cannot be modified after creation.

Mutable	                   | Immutable
---------------------------|---------------
list	                   | int
dict	                   | float
set	                       | bool
Most user-defined objects  | str
                           | tuple (its structure is immutable)
                           
A tuple can contain a mutable object, such as a list. The tuple's references 
cannot be reassigned, but the contained list may still be modified.
```

```python
'''
Object References
In Python, variables are names bound to objects, 
rather than boxes that permanently contain values.

An assignment such as:
a = [10, 20, 30]
b = a
does not create a second list. It makes b refer to the same list as a.
'''
```

```python
# Checking object identity
a = [10, 20, 30]
b = a
c = [10, 20, 30]

print("a == b:", a == b) # == compares values for equality.
print("a is b:", a is b) # 'is' checks whether two references point to the same object.

print('a == c:', a == c)
print('a is c:', a is c)
```

```python
# Passing an object to a function

def update_marks(marks): 
    marks.append(95) 
    print("Original marks before appending:",marks)
student_marks = [80, 85, 90]  
update_marks(student_marks) 
print("Updated marks after appending:",student_marks)

# The function parameter marks refers to the same list object as student_marks.
# Therefore, appending an item inside the function changes the original list.
```

```python
# However, rebinding the parameter is different

def replace_marks(marks):
    marks = [100, 100, 100]
    print("local variable marks:", marks)

student_marks = [80, 85, 90, 89]
replace_marks(student_marks)
print("list object as argument:", student_marks)

#The assignment inside replace_marks() changes only the local name marks; it does not rebind student_marks.
```

```python
'''
Introspection and Reflection
Python provides several built-in functions that allow a program to 
inspect objects, discover their attributes, examine their types, 
and modify their attributes at runtime. These capabilities are useful 
in debugging, dynamic programming, framework development, and automation.

Introspection (Passive Examination)
It is the ability of a program to examine information about itself and 
the objects it uses while it is running.

For example, a program can discover:
- The type of an object.
- The attributes and methods available in an object.
- Whether a particular attribute exists.
- The value of an attribute.

Reflection (Active Modification)
It is the ability of a program to inspect and interact with its own structure 
dynamically, including accessing or modifying attributes and invoking methods at runtime.

Introspection and reflection are closely related. 
Introspection is the passive act of examining an object's type, attributes, and metadata at runtime, 
while reflection is the active process of dynamically modifying an object's structure or behavior.
'''
```

```python
'''
getattr() Function: 
Dynamically fetches the value of an attribute by its string name.

Syntax: 
getattr(object, attribute_name)

getattr(object, attribute_name, default)
The default is returned if the attribute does not exist.
'''

class Student:
    def __init__(self, name, marks, course):
        self.name = name
        self.marks = marks
        self.course = course

student = Student("Lasya", 85, "Int. MCA")

print("student name:", getattr(student, "name")) # is equivalent print(student.name)
print("marks scored:", getattr(student, "marks"))
print("course pursed:", getattr(student, "course"))

field = input("Enter field to display: ")
value = getattr(student, field, "Field not available") # Using a default value
print(value)
```

```python
'''
set() Function:
Dynamically assigns, creates, or overwrites an attribute value, or 
attaches a new method to an object/class.

Syntax: setattr(obj, name, value)
'''

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

student = Student("Charles", 75)
print("Marks before update:", student.marks) 
setattr(student, "marks", 95) # Update an existing attribute
print("Marks after update:", student.marks)

setattr(student, "course", "MCA") # Create a new attribute
print("Course:", student.course)
print("Student has attribute course:", hasattr(student, "course"))

field = input("Enter attribute name: ")
value = input("Enter attribute value: ")
setattr(student, field, value)
print("Updated value:", getattr(student, field))
print(f"Student has attribute {field}:", hasattr(student, field))
```
