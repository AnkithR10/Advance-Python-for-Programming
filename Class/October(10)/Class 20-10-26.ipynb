# Regular Expressions
It is a sequence of characters that defines a pattern used to search, validate,extract or replace text.

It can check whether a string contains the specified search pattern

Uses:
- Valid email and phone numbers
- Extract dates,numbers and identifiers from text
- Search specified words from doc
- Replace unwanted char ot text
- Cleaning CSv data and processing log files
