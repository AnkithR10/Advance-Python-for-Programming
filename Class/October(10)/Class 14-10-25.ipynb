# Introspection and Reflection
- Inrtospection is the ability of a program to examine info about itself and the objects it uses while it is running
- Reflection is the ability of a program to inspect and interact with its own structure dynamically including accessing or modifying attributes and invoking methods at runtime.

## Built in Inspection Methods
- id(variable)
- type(variable)
- dir()
- isinstance(variable,type)
- issubclass(sub,super)
- hasattr(obj,attribute_name)
- callable(obj)
- help(obj)
- globals() and locals()

## Built in reflection methods
- getattr(object,attribute_name,default)
- setattr(obj,name,value)
- delattr(obj,attribute_name)
