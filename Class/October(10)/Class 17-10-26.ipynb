# shutil module
supports high level file operations such as copy,move and delete

## Functions
- shutil.copy() - copies a file content and permisson mode
- shutil.copy2() - copies a file and attempts to preserve additional metadata
- shutil.copytree() - copies a directory tree
- shutil.move() - moves or renames a file or directory
