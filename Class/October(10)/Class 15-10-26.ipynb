# Binary file processing
Manipilates eaw bytes exactly as they sit on the disk,bypassing character encoding.
python supports teo types of data : Text, Binary

```python
with open("C:/Degree/MCA/Sem 3/Python/download 1.png","rb") as source_file:
    with open("C:/Degree/MCA/Sem 3/Python/download 2.png","wb") as destination_file:
        chunk_count=1
        while chunk:=source_file.read(4096):
            destination_file.write(chunk)
            print(f"Writing chunk {chunk_count}")
            print(f"---Chunk {chunk_count} (length: {len(chunk)}bytes) ")
            print(f"Byte Preview: {chunk[:50]}")
            print("-"*50)
            chunk_count += 1
            
print("\nFile copy complete! Total chunks written:", chunk_count - 1)
```
