```python
N=int(input("Enter the number of elements: "))
arr=[]
for i in range(N):
    ele=int(input("Enter the element: "))
    arr.append(ele)
print("The array is: ", arr)
```

```python
#Bubble sort
arryb=arr.copy()
for i in range(N):
    for j in range(0, N-i-1):
        if arryb[j] > arryb[j+1]:
            arryb[j], arryb[j+1] = arryb[j+1], arryb[j]
print("Sorted array using Bubble sort is: ", arryb)
```

```python
#Insertion sort
arryi=arr.copy()
for i in range(1, N):
    key=arryi[i]
    j=i-1
    while j>=0 and key<arryi[j]:
        arryi[j+1]=arryi[j]
        j-=1
    arryi[j+1]=key
print("Sorted array using Insertion sort is: ", arryi)
```

```python
#Selection sort
arrys=arr.copy()
for i in range(N):
    min_idx=i
    for j in range(i+1, N):
        if arrys[j] < arrys[min_idx]:
            min_idx=j
    arrys[i], arrys[min_idx] = arrys[min_idx], arrys[i]
print("Sorted array using Selection sort is: ", arrys)
```

```python
#Merge sort
arrym=arr.copy()
for i in range(1, N):
    key=arrym[i]
    j=i-1
    while j>=0 and key<arrym[j]:
        arrym[j+1]=arrym[j]
        j-=1
    arrym[j+1]=key
print("Sorted array using Insertion sort is: ", arrym)
```

```python
#Linear search
arryl=arr.copy()
x=int(input("Enter the element to be searched: "))
for i in range(N):
    if arryl[i]==x:
        print("Element found at index: ", i)
        break
else:
    print("Element not found")
```

```python
#Binary search
arryb=arr.copy()
x=int(input("Enter the element to be searched: "))
low=0
high=N-1
while low<=high:
    mid=(low+high)//2
    if arryb[mid]==x:
        print("Element found at index: ", mid)
        break
    elif arryb[mid]<x:
        low=mid+1
    else:
        high=mid-1
else:
    print("Element not found")
```
