```python
'''
Text File Processing
Reading large files efficiently in Python requires a firm understanding 
of memory management, character encoding, and I/O buffering.

1. Reading Large Files
When a text file contains thousands or millions of lines, reading it all 
into memory can consume a lot of RAM.

Method	                | Description	                           | Memory usage
------------------------|------------------------------------------|---------------------------
read()	                | Reads the entire file by default	       | Can be high
read(size)	            | Reads a specified number of characters   | Bounded by requested size
readline()	            | Reads one line at a time	               | Usually low
readlines()	            | Reads all lines into a list	           | Can be high
Iterating over the file	| Processes lines one at a time	           | Usually low
'''
```

```python
# Reading an entire file

with open("students.txt", "r", encoding="utf-8") as file:
    content = file.read()
print(content)

'''
Here, read() loads the entire file's text into a string.
This is convenient for small files but can consume substantial memory for very large files.
'''
```

```python
'''
2. Memory-Efficient Line-by-Line Processing:
Loading an entire large file (e.g., several gigabytes) into memory using 
file.read() or file.readlines() can instantly trigger memory allocation 
errors or severe performance degradation.

The standard, most memory-efficient way to handle files of any size is 
to iterate directly over the file object. A Python file object acts as 
a lazy iterator; it reads chunks of data from disk into a small internal 
buffer and yields only one line at a time.
'''
```

```python
# Reading a large file in chunks

with open("large.txt", "r", encoding="utf-8") as file:
    while True:
        chunk = file.read(10)
        if not chunk:
            break
        print(chunk, end="\n")
# A chunk is not necessarily a complete line. A line can be split across multiple chunks.
```

```python
# Efficient line-by-line reading

with open("large.txt", "r", encoding="utf-8") as file:
    count = 0

    for line in file:
        count += 1

print("Total records:", count)
```

```python
'''
3. Serialization and deserialization
Serialization converts an object into a representation that can 
be stored or transmitted.
Deserialization reconstructs the object from that representation.

Python object
Example: a dictionary containing student details

Serialization
Convert an object to bytes or a storable representation

File or data stream
Store or transmit the representation

Deserialization
Reconstruct the Python object

Serialization is useful in:
- Saving student records between program executions.
- Storing application settings.
- Transferring data between systems.
- Saving model objects or application state.
- Caching computation results.

Common serialization approaches
Method	    | Representation	                    | Typical use
------------|---------------------------------------|----------------------------------
pickle	    | Python-specific binary representation	| Python objects and data structures
JSON	    | Text	                                | Interoperable data exchange
CSV	        | Text	                                | Tabular records
struct	    | Packed binary data	                | Fixed-format numeric or low-level data

Not every serialization format can represent every Python object. 

For example, JSON supports common data types such as dictionaries, 
lists, strings, numbers, Booleans, and None, but it does not 
directly preserve arbitrary Python class instances.

4. The pickle module
Python's built-in pickle module serializes and deserializes many Python objects, 
including dictionaries, lists, tuples, and instances of user-defined classes.

Function	             | Purpose
--------------------------------------------------------
pickle.dump(obj, file)	 | Serializes an object and writes it to a file
pickle.load(file)	     | Reads and deserializes an object from a file
pickle.dumps(obj)	     | Serializes an object and returns bytes
pickle.loads(data)	     | Deserializes an object from bytes
```

```python
# Serializing and deserializing a student dictionary

import pickle
student = {"roll_no": 101,"name": "Ananya","marks": [85, 90, 92]}

with open("student.dat", "wb") as file: # Serialization
    pickle.dump(student, file) # converts the dictionary into a serialized representation and writes it to file
print("Student record saved")

with open("student.dat", "rb") as file: # Deserialization - Reads and deserializes an object from a file
    restored_student = pickle.load(file)

print("Restored record:", restored_student)
print("Student name:", restored_student["name"])
print("Total marks:", sum(restored_student["marks"]))
```

```python
'''
Pickling a user-defined class object

Develop a menu-driven program that allows users to:
1. Add a student record.
2. Display all student records.
3. Search for a student using a roll number.
4. Save records in a binary file using pickle.
'''

import pickle
import os

FILENAME = "student_records.pkl" # .pkl - Python Pickle files
def load_records():
    if os.path.exists(FILENAME):
        with open(FILENAME, "rb") as file:
            return pickle.load(file)
    return []

def save_records(records):
    with open(FILENAME, "wb") as file:
        pickle.dump(records, file)

def add_student(records):
    roll_no = input("Enter roll number: ").strip()
    if any(s["roll_no"] == roll_no for s in records):
        print("Roll number already exists")
        return
    name = input("Enter name: ").strip()
    marks = float(input("Enter marks: "))
    records.append({"roll_no": roll_no,"name": name,"marks": marks})
    save_records(records)
    print("Student record saved")

def display_students(records):
    if not records:
        print("No records available")
        return
    for student in records:
        print(student)

def search_student(records):
    roll_no = input("Enter roll number to search: ").strip()
    for student in records:
        if student["roll_no"] == roll_no:
            print("Record found:", student)
            return
    print("Student not found")

records = load_records()
while True:
    print("\n1. Add student")
    print("2. Display students")
    print("3. Search student")
    print("4. Exit")

    choice = input("Enter choice: ").strip()
    if choice == "1":
        add_student(records)
    elif choice == "2":
        display_students(records)
    elif choice == "3":
        search_student(records)
    elif choice == "4":
        print("Exiting program")
        break
    else:
        print("Invalid choice")
```

```python
'''
Data cleaning
Data cleaning is the process of identifying and correcting missing, 
inconsistent, duplicated, or invalid data before analysis.

Problems include:
- Extra spaces in names.
- Missing marks for Ravi.
- Marks exceeding the allowed range of 0–100.
- A duplicate record for Meera.

Problem	Cleaning operation
Extra spaces	    | strip()
Missing values	    | Detect blanks and handle them appropriately
Duplicate records	| Track unique keys or compare rows
Invalid numbers	    | Use try/except during conversion
Out-of-range marks	| Validate against allowed limits
Inconsistent case	| Apply lower() or upper() when appropriate
Missing columns	    | Check expected headers before processing
'''
```

```python
import csv
import json
students = [
    {"RollNo ": 101, " Name": "Anu", "Python_Marks": 90, "DBMS_Marks": 80},
    {"RollNo": 102, "Name": "Ravi", "Python_Marks": 85, "DBMS_Marks": 90},
    {"RollNo": 103, "Name": "Meera", "Python_Marks": "", "DBMS_Marks": 70},
    {"RollNo": 103, "Name": "Meera", "Python_Marks": 95, "DBMS_Marks": 70}
]
fields = ["RollNo", "Name", "Python_Marks", "DBMS_Marks"]
with open("student_report.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=fields)
    writer.writeheader()
    writer.writerows(students)
print("Student records written to CSV")
cleaned_students = []
seen_roll_numbers = set()
with open("student_report.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        roll_no = (row["RollNo"] or "").strip()
        name = (row["Name"] or "").strip()
        python_text = (row["Python_Marks"] or "").strip()
        dbms_text = (row["DBMS_Marks"] or "").strip()

        if not roll_no or not name: # Skip missing identification details
            continue

        if roll_no in seen_roll_numbers: # Remove duplicate roll numbers
            print("Duplicate skipped:", roll_no)
            continue

        try:  # Handle missing marks using the subject average
            dbms_marks = float(dbms_text)
            if not 0 <= dbms_marks <= 100:
                continue
            if python_text:
                python_marks = float(python_text)
                if not 0 <= python_marks <= 100:
                    continue
            else:
                python_marks = 90  # Replace with a justified imputation value
        except ValueError:
            print("Invalid marks skipped:", roll_no)
            continue
        total = python_marks + dbms_marks
        average = total / 2
        seen_roll_numbers.add(roll_no)
        cleaned_students.append({
            "RollNo": roll_no,
            "Name": name,
            "Python_Marks": python_marks,
            "DBMS_Marks": dbms_marks,
            "Total": total,
            "Average": average
        })

# Step 3: Save cleaned data to CSV
report_fields = ["RollNo", "Name", "Python_Marks", "DBMS_Marks", "Total", "Average"]

with open("cleaned_student_report.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=report_fields)
    writer.writeheader()
    writer.writerows(cleaned_students)

# Step 4: Save cleaned data to JSON
with open("student_report.json", "w", encoding="utf-8") as file:
    json.dump(cleaned_students, file, indent=4)

# Step 5: Display results
print("\nCleaned student report:")
for student in cleaned_students:
    print(student["Name"],"Total:", student["Total"],"Average:", student["Average"])
print("\nValid students:", len(cleaned_students))
print("CSV and JSON reports saved")
```

```python
'''
Regular Expressions
A regular expression is a sequence of characters that defines a 
pattern used to search, validate, extract, or replace text.

RegEx can check whether a string contains the specified search pattern.

Regular expressions are useful for:
- Validating email addresses and phone numbers.
- Extracting dates, numbers, and identifiers from text.
- Searching for specific words in documents.
- Replacing unwanted characters or text.
- Cleaning CSV data and processing log files.

For example, consider this text:
Student Anu scored 85 marks. Student Ravi scored 92 marks.

Using regular expressions, we can extract all the marks (85, 92) 
without manually checking each word.

The re module
Python provides the built-in re module for regular-expression operations.

Import the re module:
import re

Common functions
Function	   | Purpose
---------------|---------------------------------------
re.search()	   | Finds the first match anywhere in a string
re.match()	   | Checks for a match only at the beginning of a string
re.fullmatch() | Checks whether the entire string matches the pattern
re.findall()   | Extracts all non-overlapping matches into a list
re.finditer()  | Returns an iterator of match objects
re.sub()	   | Replaces matching text
re.split()	   | Splits text using a matching pattern

Basic regular-expression symbols/ Metacharacters
Metacharacters are characters with special meaning; they form the 
foundation of regex patterns.

Symbol	|Meaning	                                    | Example pattern	| Matches
--------|-----------------------------------------------|-------------------|------------------------
.	    | Any character except newline by default	    | a.c               | abc, a7c
^	    | Beginning of string	                        | ^Hello	        | Hello world
$	    | End of string	                                | end$	            | the end
*	    | Zero or more repetitions	                    | ab*	            | a, ab, abb
+	    | One or more repetitions	                    | ab+	            | ab, abb
?	    | Zero or one repetition	                    | colou?r	        | color, colour
\       | Signals a special or escape sequence          | \bin              | spain, ink
\d	    | A digit	                                    | \d+	            | 123
\w	    | A word character	                            | \w+	            | student_1
\s	    | Whitespace	                                | \s+	            | Spaces, tabs
[abc]	| One character from a set	                    | [aeiou]	        | Any listed vowel
{n}	    | Exactly n repetitions	                        | \d{4}	            | Four digits
{m,n}	| Between m and n repetitions	                | \d{2,4}	        | Two to four digits
()	    | Capturing group	                            | (\d{2})-(\d{4})	| Captures the two parts
Tip: Write patterns as raw strings, such as r"\d+", to avoid unintended interpretation of backslashes 
by Python string literals.

Special Sequences
A special sequence is a \ followed by one of the characters in the list. 
Character	Description	                                                Example	
\A	        Matches at the start of the string.                         "\AThe"
\b	        Matches a word boundary (beginning or end).                 r"\bain", r"ain\b" 
\B	        Matches where there is no word boundary.                    r"\Bain", r"ain\B"
\d	        Matches a digit (0–9).                                      "\d"
\D	        Matches a non-digit.                                        "\D"
\s	        Matches a whitespace character.                             "\s"
\S	        Matches a non-whitespace character.                         "\S"
\w	        Matches a word character (letter, digit, or underscore).    "\w"
\W	        Matches a non-word character.                               "\W"
\Z	        Matches at the end of the string.                           "\Z"

Sets
A set is a set of characters inside a pair of square brackets [] with a special meaning:

Set	         Description	
[arn]	     Returns a match where one of the specified characters (a, r, or n) is present	
[a-n]	     Returns a match for any lower case character, alphabetically between a and n	
[^arn]	     Returns a match for any character EXCEPT a, r, and n	
[0123]	     Returns a match where any of the specified digits (0, 1, 2, or 3) are present	
[0-9]	     Returns a match for any digit between 0 and 9	
[0-5][0-9]	 Returns a match for any two-digit numbers from 00 and 59	
[a-zA-Z]	 Returns a match for any character alphabetically between a and z, lower case OR upper case	
[+]	         In sets, +, *, ., |, (), $,{} has no special meaning, so [+] means: return a match for any + character in the string

Flags
It is used to add flags to the pattern when using regular expressions.

Flag	        Shorthand	Description	Try it
re.ASCII	    re.A	    Returns only ASCII matches	
re.DEBUG		            Returns debug information	
re.DOTALL	    re.S	    Makes the . character match all characters (including newline character)	
re.IGNORECASE	re.I	    Case-insensitive matching	
re.MULTILINE	re.M	    Returns matches at the start/end of each line	
re.NOFLAG		            Specifies that no flag is set for this pattern	
re.UNICODE	    re.U	    Returns Unicode matches. This is the default in Python 3.	
re.VERBOSE	    re.X	     Allows whitespace and comments inside patterns. Makes the pattern more readable
```

```python
#  Metacharacters
import re

tests = [
    (r"c.t", "The cat slept"),                 # . any character
    (r"^Namah", "Namah Shivaya"),              # ^ start of string
    (r"night$", "Good night"),                 # $ end of string
    (r"bon*", "bonjour!"),                     # * zero or more
    (r"go+", "gooo!"),                         # + one or more
    (r"behaviou?r", "behavior"),               # ? zero or one
    (r"\bfox", "A fox ran"),                   # \b word boundary
    (r"567\b", "Order no: 567"),                   # \b word boundary
    (r"\d+", "Room 508"),                      # \d digit
    (r"\w+", "guest42"),                       # \w word characters
    (r"\s+", "red blue"),                      # \s whitespace
    (r"[A-Z]", "Python"),                      # one character from a set
    (r"\d{3}", "Code 729"),                    # exactly 3 repetitions
    (r"\d{2,3}", "Age 25"),                    # 2 to 3 repetitions
    (r"([A-Z]{2})-(\d{3})", "ID AB-729"),      # capturing groups
]

for pattern, text in tests:
    match = re.search(pattern, text)
    if match:
        print(f"{pattern} in {text} → {match.group()}")
        if match.groups():
            print(f"Captured groups: {match.groups()}")
    else:
        print(f"{pattern} in {text} → no match")
```

```python
'''
findall() vs finditer()
Feature	           | re.findall()	                                       | re.finditer()
Return Type	       | A standard list	                                   | A lazy iterator
Element Type	   | Strings (or tuples of strings if groups are used)     | Match objects
Memory Footprint   | High (loads all matches into memory at once)	       | Low (processes matches one by one on demand)
Metadata Access	   | No (loses positions, indexes, and span data)	       | Yes (gives access to .start(), .end(), and .span())
Best Used For	   | Small strings where you only need the text quickly    | Large datasets, logs, or when you need match positions
```

```python
import re

txt = "Ånand"

# Find all ASCII matches
print("ASCII:",re.findall("\w", txt, re.ASCII))

# Without the flag, the example would return all characters
print("raw string:", re.findall("\w", txt))

# Same result using the shorthand re.A flag
print("shorthand ASCII:", re.findall("\w", txt, re.A))

# Find all UNICODE matches
print("UNICODE:", re.findall("\w", txt, re.UNICODE))

# Search for a case-insensitive match
print("DEBUG:", re.findall("an", txt, re.DEBUG)) 
```

```python
import re

txt = """Hi
my
name
is
Chethana"""

#Search for a sequence that starts with "me", followed by one character, even a newline character, and continues with "is"
print(re.findall("me.is", txt, re.DOTALL))

#This example would return no match without the re.DOTALL flag
print(re.findall("me.is", txt))

#Same result with the shorthand re.S flag
print(re.findall("me.is", txt, re.S))

#Search for the sequence "na", at the beginning of a line
print(re.findall("^na", txt, re.MULTILINE))

#This example would return no matches without the re.MULTILINE flag
print(re.findall("^na", txt))

#Same result with the shorthand re.M flag
print(re.findall("^na", txt, re.M))
```

```python
import re

text = "The rain in Mysuru falls mainly on the city area rather than farm"
standard_pattern = r"\b\w*ain\w*\b" 
print("Standard:", re.findall(standard_pattern, text))

# Whitespace and comments are ignored by the engine
verbose_pattern = r"""
    \b        # Start at a word boundary
    \w*       # Match any leading letters of the word
    ain       # Match the literal phrase 'ain'
    \w*       # Match any trailing letters of the word
    \b        # End at a word boundary
"""
print("Verbose (re.VERBOSE):", re.findall(verbose_pattern, text, re.VERBOSE))
print("Verbose (re.X):", re.findall(verbose_pattern, text, re.X))

# It looks for actual spaces and newlines in the text, so it finds nothing.
print("Verbose pattern without flag:", re.findall(verbose_pattern, text))
```

```python
'''
Compiled Regular Expression
It allows you to compile a regular expression pattern into a RegEx object.
This compiled object can then be reused for multiple operations like search, etc.
'''
import re
regex = re.compile(r'([\d]{2})-([\d]{2})-([\d]{4})') 
print('compiled reg expr:', regex.search('26-10-2026'))
print(regex.sub(r'\1.\2.\3', '27-10-2026')) # (\1) Group 1, (\2) Group 2,(\3) Group 3
```

```python
'''
Log File Analyzer

What is a Log File?
A log file is a file that stores records of events generated by a computer system, 
application, server, or network device. These records help developers and system 
administrators understand system behavior, identify errors, monitor performance, 
and investigate security incidents.

A log file may contain:
- Timestamps: Date and time when an event occurred.
- Log levels: INFO, WARNING, ERROR, DEBUG, or CRITICAL.
- Source information: Name of the application, module, or service.
- Event messages: Details about the activity or error.
- Error codes: Codes that help identify specific problems.

Sample Log File
Consider the following log file named application.log:
2026-10-24 09:15:23 INFO User login successful: user_id=101
2026-10-24 09:16:10 WARNING Disk space is running low: 15% remaining
2026-10-24 09:17:45 ERROR Database connection failed: error_code=DB101
2026-10-24 09:18:02 INFO Database connection restored
2026-10-24 09:19:30 ERROR File not found: report.pdf
2026-10-24 09:20:15 CRITICAL Server is not responding
2026-10-24 09:21:00 INFO User logout successful: user_id=101

Manually inspecting hundreds or thousands of log entries can be time-consuming. 
Python's re module can automate this process by extracting timestamps, identifying 
log levels, searching for patterns, and detecting errors.
'''
```

```python
# read a log file named application.log and display its contents line by line.
with open("application.log", "r", encoding="utf-8") as file:
    for line in file:
        print(line.rstrip())
```

```python
# Timestamp Extraction - to extract all timestamps from a log file
import re

pattern = re.compile(r"\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}")
with open("application.log", "r", encoding="utf-8") as file:
    content = file.read()
timestamps = pattern.findall(content)

print("Extracted Timestamps:")
for time in timestamps:
    print(time)
print("Total timestamps:", len(timestamps))
```

```python
# Log Parsing and Pattern Matching
import re

pattern = re.compile(
    r"^(\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2})"
    r"\s+(INFO|WARNING|ERROR|CRITICAL|DEBUG)\s+(.*)$"
)
with open("application.log", "r", encoding="utf-8") as file:
    for line in file:
        line = line.rstrip("\n")
        match = pattern.match(line)
        if match:
            timestamp = match.group(1)
            level = match.group(2)
            message = match.group(3)
            print("Timestamp:", timestamp)
            print("Level:", level)
            print("Message:", message)
            print("-" * 50)
        else:
            print("Unrecognized log entry:", line)
```

```python
# Error Detection
import re

error_pattern = re.compile(r"\b(ERROR|CRITICAL)\b")
error_count = 0
critical_count = 0

with open("application.log", "r", encoding="utf-8") as file:
    for line in file:
        match = error_pattern.search(line)
        if match:
            level = match.group(1)
            print(line.rstrip())
            if level == "ERROR":
                error_count += 1
            elif level == "CRITICAL":
                critical_count += 1
print("\nError Summary")
print("ERROR entries:", error_count)
print("CRITICAL entries:", critical_count)
print("Total detected entries:", error_count + critical_count)
```

```python
# Extract Error Codes
import re

pattern = re.compile(r"error_code=(\w+)")
with open("application.log", "r", encoding="utf-8") as file:
    for line in file:
        codes = pattern.findall(line)
        if codes:
            print("Log:", line.rstrip())
            print("Error codes:", ", ".join(codes))
            print()
```

```python
# Detect Invalid Log Entries
import re

pattern = re.compile(
    r"^\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}"
    r"\s+(INFO|WARNING|ERROR|CRITICAL|DEBUG)\s+.+$"
)

with open("application.log", "r", encoding="utf-8") as file:
    for line_number, line in enumerate(file, start=1):
        line = line.rstrip("\n")
        if pattern.fullmatch(line):
            print(f"Line {line_number}: Valid")
        else:
            print(f"Line {line_number}: Invalid - {line}")
```

```python
import re
from collections import Counter
from datetime import datetime

log_pattern = re.compile( # Captures the message, including spaces, after the log level
    r"^(\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2})"
    r"\s+(INFO|WARNING|ERROR|CRITICAL|DEBUG)\s+(.*)$" 
)

counts = Counter()
timestamps = []
error_messages = []
total_entries = 0
invalid_entries = 0

with open("application.log", "r", encoding="utf-8") as file:
    for line_number, line in enumerate(file, start=1):
        line = line.rstrip("\n")
        if not line.strip():
            continue
            
        match = log_pattern.fullmatch(line)
        
        if not match:
            invalid_entries += 1
            continue
        timestamp_text = match.group(1)
        level = match.group(2)
        message = match.group(3)
        
        try:
            timestamp = datetime.strptime(timestamp_text, "%Y-%m-%d %H:%M:%S")
        except ValueError:
            invalid_entries += 1
            continue
            
        total_entries += 1
        counts[level] += 1
        timestamps.append(timestamp)

        if level in ("ERROR", "CRITICAL"):
            error_messages.append(f"{timestamp_text} {level} {message}")

with open("log_report.txt", "w", encoding="utf-8") as report:
    report.write("APPLICATION LOG ANALYSIS REPORT\n")
    report.write("=" * 40 + "\n\n")

    report.write(f"Total valid entries: {total_entries}\n")
    report.write(f"Invalid entries: {invalid_entries}\n\n")

    report.write("LOG LEVEL SUMMARY\n")
    report.write("-" * 40 + "\n")

    for level in ["INFO", "WARNING", "ERROR", "CRITICAL", "DEBUG"]:
        report.write(f"{level}: {counts[level]}\n")

    report.write("\nERROR AND CRITICAL MESSAGES\n")
    report.write("-" * 40 + "\n")

    if error_messages:
        for message in error_messages:
            report.write(message + "\n")
    else:
        report.write("No errors or critical events found.\n")

    report.write("\nTIMESTAMP SUMMARY\n")
    report.write("-" * 40 + "\n")

    if timestamps:
        report.write(f"Earliest timestamp: {min(timestamps)}\n")
        report.write(f"Latest timestamp: {max(timestamps)}\n")
    else:
        report.write("No valid timestamps found.\n")

print("Log analysis completed.")
print("Report saved as log_report.txt")
```
