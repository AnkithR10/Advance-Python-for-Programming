```python
# recursion cannot be always considered to get the efficient solution-->stack memory
#Sol--->Memorisation
def sum_of_n_numbers(n):
    if n<=0:  
        return 0
    return n+sum_of_n_numbers(n-1)


n=int(input("Enter the n"))
print("Sum is---:", sum_of_n_numbers(n))
```

```python
''' Memorization to optimize recursion

'''  
# Fibonacci series using recursion memorization using decorrator
from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
print(fibonacci(10))  # Output: 55
```
