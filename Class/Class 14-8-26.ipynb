```python
# while loop with product incrementation
n=int(input())
product = 1
i = n
while i > 0:
    product *= i
    i -= 1
print(product)
```

```python
n=int (input())
product = 1
i = n
while i > 0:
    product *= i      
    product /= i     
    i -= 1

print(product)
```

```python
#Calculate Area of a circle
def calculate_area():
    radius = float(input("Enter radius : "))
    area = 3.14159 * radius ** 2   
    print("Area:", area)
calculate_area()
```

```python
#Return value
#Functions can send data back to the code that called them
def greet(name):
    return "Hello, " + name + "!"
print(greet("Ankith"))
```

```python
#Default Arguments
def add(a,b=0):
    print("Sum=",a+b)

add(5,10)
```

```python
#Default Arguments help to solve Mismatch of parameters
def student(name,message="Welcome to Amrita",course="MCA"):
    print("Hello",name)
    print(message)
    print("Course:",course)

student("Ankith")
print()
student("Ankith","Welcome to Amrita University")
```
