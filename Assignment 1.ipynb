```python
# 1. Increasing Right Triangle
n = int(input())
for i in range(1, n + 1):
    print("*" * i)
```

```python
# 2. Decreasing Right Triangle
n = int(input())
for i in range(n, 0, -1):
    print("*" * i)
```

```python
# 3. Right-Aligned Increasing Triangle
n = int(input())
for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)
```

```python
# 4. Right-Aligned Decreasing Triangle
n = int(input())
for i in range(n, 0, -1):
    print(" " * (n - i) + "* " * i)
```

```python
# 5. Full Pyramid
n = int(input())
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
# 6. Inverted Pyramid
n = int(input())
for i in range(n, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
# 7. Diamond
n = int(input())
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
# 8. Hollow Pyramid
n = int(input())
for i in range(1, n + 1):
    if i == n:
        print("*" * (2 * n - 1))
    elif i == 1:
        print(" " * (n - 1) + "*")
    else:
        print(" " * (n - i) + "*" + " " * (2 * i - 3) + "*")
```

```python
# 9. Solid Square
n = int(input())
for i in range(n):
    print("* " * n)
```

```python
# 10. Hollow Square
n = int(input())
for i in range(n):
    if i == 0 or i == n - 1:
        print("* " * n)
    else:
        print("* " + "  " * (n - 2) + "*")
```

```python
# 11. Number Square
n = int(input())
for i in range(n):
    print(*range(1, n + 1))
```

```python
# 12. Same Number Square
n = int(input())
for i in range(1, n + 1):
    print(*([i] * n))
```

```python
# 13. Increasing Number Triangle
n = int(input())
for i in range(1, n + 1):
    print(*range(1, i + 1))
```

```python
# 14. Decreasing Number Triangle
n = int(input())
for i in range(n, 0, -1):
    print(*range(1, i + 1))
```

```python
# 15. Repeated Number Triangle
n = int(input())
for i in range(1, n + 1):
    print(*([i] * i))
```

```python
# 16. Reverse Number Triangle
n = int(input())
for i in range(1, n + 1):
    print(*range(n, n - i, -1))
```

```python
# 17. Continuous Number Triangle
n = int(input())
x = 1
for i in range(1, n + 1):
    for j in range(i):
        print(x, end=" ")
        x += 1
    print()
```

```python
# 18. Number Pyramid Pattern
n = int(input())
for i in range(1, n + 1):
    print(" " * (n - i), end="")
    print(*range(1, i + 1), *range(i - 1, 0, -1))
```

```python
# 19. Increasing Alphabet Triangle
n = int(input())
for i in range(1, n + 1):
    for j in range(i):
        print(chr(65 + j), end=" ")
    print()
```

```python
# 20. Repeated Alphabet Triangle
n = int(input())
for i in range(n):
    print(*([chr(65 + i)] * (i + 1)))
```

```python
# 21. Continuous Alphabet Triangle
n = int(input())
x = 65
for i in range(1, n + 1):
    for j in range(i):
        print(chr(x), end=" ")
        x += 1
    print()
```

```python
# 22. Reverse Alphabet Triangle
n = int(input())
for i in range(n, 0, -1):
    for j in range(i):
        print(chr(65 + j), end=" ")
    print()
```

```python
# 23. X Pattern
n = int(input())
for i in range(n):
    for j in range(n):
        if j == i or j == n - i - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# 24. Plus Pattern
n = int(input())
m = n // 2
for i in range(n):
    for j in range(n):
        if i == m or j == m:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# 25. Hollow Diamond
n = int(input())
for i in range(1, n + 1):
    if i == 1:
        print(" " * (n - 1) + "*")
    else:
        print(" " * (n - i) + "*" + " " * (2 * i - 3) + "*")
for i in range(n - 1, 0, -1):
    if i == 1:
        print(" " * (n - 1) + "*")
    else:
        print(" " * (n - i) + "*" + " " * (2 * i - 3) + "*")
```

```python
# 26. Hourglass Pattern
n = int(input())
for i in range(n, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
for i in range(2, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
# 27. Filtering based on Condition
numbers = [12, 45, 7, 23, 56, 18, 90]
large_numbers = [x for x in numbers if x > 21]
print(large_numbers)
```

```python
# 28. Reversing a String with For Loop
original_str = "Python"
reversed_str = ""
for char in original_str:
    reversed_str = char + reversed_str
print(reversed_str)
```
